## A note on notation used throughout

- Bold or arrow-marked letters (**v**, $\vec{v}$) are vectors — they have both size and direction.
- Plain letters (v, m, t) are scalars — just a size.
- $\Delta$ (delta) always means "change in" — final minus initial.
- Subscript 0 (like $v_0$) always means "at the start" (initial value).
- We'll almost always pick a positive direction explicitly before solving any vector problem — and once picked, it never changes mid-problem.

# UNIT 1 — Kinematics
### *(Handouts, Classes 1–2)*

Kinematics is the study of motion itself — position, velocity, acceleration, and how they relate — **without asking what caused the motion.** That question (what forces are involved) is next unit, Dynamics. For now, we're just building a vocabulary and a toolkit for describing *how* something moves.

---

## 1.1 Position — the starting point for everything

**The intuition.** Before you can describe motion, you need to be able to describe *where something is*. That sounds trivial, but "where" always secretly requires two things: a reference point (where are you measuring *from*?) and a direction (which way?). "5 meters" means nothing on its own — 5 meters from what, in which direction? That's what makes position a **vector**: a quantity with both a *magnitude* (the 5 meters) and a *direction* (an angle relative to some reference line you chose).

Contrast this with a **scalar** — a quantity that's fully described by a single number. Temperature, mass, time: these don't have a "direction." Position, velocity, force: these do.

**Why this distinction matters so much:** almost every mistake in this entire course, from kinematics through momentum, traces back to treating a vector like a scalar (forgetting a sign or direction) or vice versa. Get comfortable with this now and it pays off for the whole year.

**The unit.** Position is measured in meters (m) in SI units. Meters is one of the *fundamental* units — one of a small set of independently-defined units (there are seven in the SI system: meter, kilogram, second, ampere, kelvin, mole, candela) that everything else (velocity, force, energy...) is built out of combinations of. Every other unit in this course (m/s, N, J...) is a **derived unit** — some combination of the fundamental seven.

**Common mistake:** writing down a distance with no reference to direction or origin, then being surprised when a problem needs a sign. Get in the habit: whenever you write down a position, ask "position relative to *what*, and which way is positive?"

## 1.2 Displacement vs. Distance

**The intuition.** This is the single most important vector-vs-scalar pair in kinematics, and it's worth internalizing with a physical picture, not just a definition.

Imagine walking from your front door, all the way around the block, and back to your front door.

- **Distance** is what your pedometer says — it only ever adds up. Every step forward, however winding your path, adds to the total. It is a scalar; it cannot be negative, and it cannot "cancel out."
- **Displacement** is the straight arrow from where you started to where you ended up. It's $\Delta \vec{r} = \vec{r}_f - \vec{r}_i$ — a vector, with both size and direction. If you end up back at your front door, your displacement is **exactly zero**, no matter how far you walked.

This is genuinely counterintuitive the first time you see it, because in everyday speech we use "how far did you go" to mean distance. In physics, "how far" is ambiguous — you have to ask "distance, or displacement?"

**Worked example 1 (straightforward):** You drive 8 km east, then 3 km west. Distance traveled = 8 + 3 = 11 km. Displacement = 8 − 3 = 5 km east (net).

**Worked example 2 (the classic gotcha — round trip):** A runner does one full lap of a circular track of radius $r$. Distance = the circumference, $2\pi r$. Displacement = 0 (same start and end point). Now suppose the runner does a **half** lap — goes from one side of the circle straight through to the diametrically opposite point. Distance = half the circumference = $\pi r$. Displacement = the straight-line diameter, $2r$ — a completely different number, using a completely different piece of geometry (a chord, not an arc).

**Common mistake:** using "distance" and "displacement" interchangeably in word problems, especially in round-trip or curved-path scenarios. If a problem describes any kind of loop, curve, or back-and-forth, stop and ask explicitly which one is being asked for.

## 1.3 Velocity vs. Speed

**The intuition.** Velocity answers "how fast is my *position* changing, and which way?" Speed answers only "how fast" — it throws away the direction information. This mirrors the displacement/distance split exactly, one derivative up: velocity is to displacement as speed is to distance.

**The equation.** Average velocity: $\vec{v} = \dfrac{\Delta \vec{r}}{\Delta t}$ (a vector, units m/s). Average speed $= \dfrac{\text{distance}}{\Delta t}$ (a scalar).

An important convention: **in physics formulas, unless a problem explicitly says "average," every quantity is understood to be *instantaneous*** — the value *at* a particular moment, not averaged over some interval. We'll come back to this distinction in 1.5.

**Worked example:** Same runner as above does one full lap (circumference $2\pi r$) in time $T$. Average speed $= \frac{2\pi r}{T}$ — a real, nonzero number. Average velocity $= \frac{0}{T} = 0$, because displacement was zero. If someone only tells you "the average velocity was 0 m/s," that does *not* mean the runner was standing still the whole time — it only tells you they ended up where they started.

**Common mistake:** plugging a distance into a velocity formula (getting a "velocity" that's actually a speed with no meaningful sign), or reporting a velocity's magnitude as if it answers a "how fast" question when the problem actually wanted the scalar speed (or vice versa). Always check: does the answer need a direction attached?

## 1.4 Acceleration

**The intuition.** Acceleration measures how fast *velocity itself* is changing. This is the step where students most often get tripped up, because "accelerating" in everyday language means "speeding up." In physics, acceleration means **any change in velocity at all** — speeding up, slowing down, *or* changing direction. All three count.

That third case — changing direction — is easy to overlook. A car going around a curve at a perfectly constant *speed* is still *accelerating*, because its velocity (which includes direction) is changing every instant. We'll come back to this in a big way in Unit 5 (circular motion), but it's worth planting the seed now: **acceleration is about velocity changing, not speed changing.**

**The equation.** $\vec{a} = \dfrac{\Delta \vec{v}}{\Delta t}$, unit m/s² (meters per second, per second — literally "how many m/s of velocity you gain, every second").

**Notice something:** there's no scalar counterpart to acceleration the way speed is the scalar counterpart to velocity. That's because acceleration is fundamentally tied to the vector velocity — there isn't a meaningful "unsigned" version of it in the same way.

**Worked example:** A ball thrown straight up decelerates on the way up (velocity and acceleration point opposite ways), momentarily has zero velocity at the peak, then accelerates back downward (velocity and acceleration now point the same way) — even though the acceleration (gravity) never changes at all throughout the whole flight. This single example is worth sitting with: **acceleration can be constant while velocity is doing three different things (decreasing, zero, increasing) over the course of the motion.**

**Common mistake:** assuming that if velocity is zero at some instant, acceleration must also be zero there. At the very peak of a thrown ball's flight, velocity = 0 but acceleration is still fully $-g$ — that's *exactly* what causes the object to immediately start moving downward again. Zero velocity is an instant; acceleration doesn't pause for it.

## 1.5 Average vs. Instantaneous

**The intuition.** "Average" velocity over some stretch of time can hide a lot of detail — a car could average 30 mph over an hour by going exactly 30 mph the whole time, or by sitting still for 30 minutes and then going 60 mph for the other 30. Both give the same average, but they're very different motions. **Instantaneous** velocity is the value at one single moment — effectively, the average velocity taken over a time interval so short it shrinks to a single instant.

**The key convention for this entire course:** unless a problem explicitly writes "average" (like $v_\text{avg}$), every symbol in a physics formula refers to the *instantaneous* value. This matters because it tells you *when* in a motion an equation's answer applies.

**Where this shows up concretely:** the kinematics equations you're about to learn (Formula 2) actually mixes an average and an instantaneous quantity in a subtle way during its derivation — worth watching for.

## 1.6 The Four Kinematics Equations — and *why* they look the way they do

**The big picture first.** For the special (but very common) case of **constant acceleration**, you can connect five quantities — $v_0$ (initial velocity), $v$ (final velocity), $a$ (acceleration), $\Delta x$ (displacement), and $t$ (time) — using just four equations, entirely with algebra (no calculus needed). Each equation deliberately **leaves out exactly one of the five quantities** — and that's the practical trick to choosing the right one fast: identify which variable you *don't* have and *don't need*, and pick the formula missing it.

**Deriving Formula 1 (leaves out $\Delta x$).** Start from the very definition of acceleration. Since acceleration is constant here, its average equals its instantaneous value:
$$a = a_\text{avg} = \frac{\Delta v}{\Delta t} = \frac{v - v_0}{t}$$
Solve for $v$:
$$\boxed{v = v_0 + at}$$
This is nothing more than "final velocity = initial velocity, plus however much you gained (or lost) over time $t$ at rate $a$."

**Deriving Formulas 2 & 3 (Formula 2 leaves out $a$; Formula 3 leaves out $v$).** Start from the definition of average velocity, $v_\text{avg} = \Delta x / \Delta t$. Here's a subtlety worth sitting with: for constant acceleration, the average velocity over the interval turns out to equal the simple average of the start and end velocities — $v_\text{avg} = \frac{1}{2}(v_0 + v)$ (this is really just "the area under a straight-line $v$-$t$ graph is a trapezoid, and a trapezoid's average height is the average of its two parallel sides"). Substituting:
$$\boxed{\Delta x = \tfrac{1}{2}(v_0 + v)t} \quad \text{(Formula 2)}$$
Now substitute Formula 1's expression for $v$ into Formula 2 to eliminate $v$ entirely:
$$\Delta x = \tfrac{1}{2}(v_0 + v_0 + at)t = \tfrac{1}{2}(2v_0 + at)t$$
$$\boxed{\Delta x = v_0 t + \tfrac{1}{2}at^2} \quad \text{(Formula 3)}$$
Formula 3 is worth an intuitive read: the first term ($v_0t$) is "how far you'd go if velocity never changed"; the second term ($\frac{1}{2}at^2$) is the *extra* distance you cover because you were also accelerating.

**Deriving Formula 4 (leaves out $t$).** Sometimes you're given $v_0$, $a$, and $\Delta x$, and asked for $v$ — but you don't know (and don't want to bother finding) $t$. Combine Formulas 1 and 3 to eliminate $t$ algebraically:
$$\boxed{v^2 = v_0^2 + 2a\Delta x}$$

**The four formulas, together:**
$$v = v_0 + at \qquad \Delta x = \tfrac12(v_0+v)t \qquad \Delta x = v_0t + \tfrac12 at^2 \qquad v^2 = v_0^2 + 2a\Delta x$$

**No more, no less.** These four are the *complete* set for constant acceleration — there's no fifth independent equation to learn. Note also: **all vector quantities here are along the same line**, so for 1-D constant-acceleration motion you can treat them like signed scalars rather than juggling full vector notation.

One bookkeeping note: many textbooks (and the AP formula sheet) write $\Delta x$ as $x_f - x_i$, sometimes shortened to $x - x_0$. It's the same $\Delta x$ — just don't let the different notation trip you up when you flip to a reference sheet.

---

### Worked Example — choosing the right formula

**Setup:** A car starts from rest and accelerates at $a = 5\ \text{m/s}^2$ for $4\text{ s}$.

**Find the velocity after 4 s.** We're given $v_0 = 0$, $a = 5$, $t = 4$ — and we're *not* given or asked about $\Delta x$. That missing-variable points straight at **Formula 1**: $v = 0 + 5(4) = 20\ \text{m/s}$.

**Find the distance covered in that time.** Now we want $\Delta x$, and we're *not* given (or don't want to separately compute) $v$ — that points at **Formula 3**: $\Delta x = 0 + \tfrac12(5)(4)^2 = 40\ \text{m}$.

**Then:** the driver sees a red light, hits the brakes, and now $a = -10\ \text{m/s}^2$ (note the sign — deceleration, opposing the direction of motion). Find the distance needed to stop. Here the new "initial" velocity is the $20\ \text{m/s}$ we just found, the final velocity is $0$ (stopped), and we want $\Delta x$ — with no interest in $t$. That's **Formula 4**: $0 = 20^2 + 2(-10)\Delta x \Rightarrow \Delta x = 20\ \text{m}$.

Notice how, at each step, the choice of formula came from asking *"what am I not given and not asked to find?"* — not from memorizing "use this formula for this type of problem."

## 1.7 Graphing Kinematics — reading slopes and areas

**The intuition.** Position, velocity, and acceleration are linked by a beautiful and totally general pattern that works regardless of whether acceleration is constant:

- **Going from $x$-graph up to $v$-graph, and $v$-graph up to $a$-graph:** take the **slope**.
- **Going from $a$-graph down to $v$-graph, and $v$-graph down to $x$-graph:** take the **area** underneath the curve.

This is really just calculus in disguise (velocity is the derivative of position; acceleration is the derivative of velocity — and integration undoes differentiation), but you don't need calculus notation to use it. You just need to be able to read a slope and estimate an area on a graph.

**Three canonical cases, worth internalizing as a set:**

| If... | then the $v$-$t$ graph is... | and the $x$-$t$ graph is... |
|---|---|---|
| $x$ = constant | flat at $v=0$ | flat |
| $v$ = constant | flat, nonzero | straight line (constant slope) |
| $a$ = constant | straight line (constant slope) | parabola |

**Worked example — the "complicated" case.** Suppose a $v$-$t$ graph shows velocity starting positive, decreasing linearly to zero, then continuing to decrease into negative territory, before leveling off. Reading this directly: the object starts moving in the positive direction, decelerates, momentarily stops, then reverses and speeds up in the negative direction, then eventually moves at a constant (negative) velocity. The **area** under the curve while it's positive is a chunk of positive displacement; the **area** while it's negative subtracts from that. The total displacement is the net (signed) area — not just "the area," since area below the axis counts as negative displacement.

**Common mistake:** treating "area under a v-t graph" as always positive, the way you might for a simple geometry area problem. When velocity is negative, that region's area represents *negative* displacement and must be subtracted, not added, from the total.

## 1.8 Free Fall

**The intuition.** Free fall isn't a new topic — it's simply the four kinematics equations from 1.6, applied to the single most common constant acceleration in this course: gravity, $a = g \approx 9.8\ \text{m/s}^2$ (often rounded to $10\ \text{m/s}^2$ for quick estimates), which **always points downward**, regardless of whether the object is moving up, down, or is momentarily at rest.

**The one skill that matters most here: picking (and holding) a sign convention.** Before plugging in a single number, decide which direction is positive — up, or down — and then never change your mind partway through the problem. A clean process:
1. Choose "up" (or "down") as positive — pick whichever makes the specific problem's numbers feel more natural.
2. Assign the *initial* velocity's sign based on that choice.
3. Write $a = -g$ (if up is positive) or $a = +g$ (if down is positive) — consistently.
4. Solve, then interpret the *sign* of your answer using the same convention.

**Worked example (downward motion) — dropping a stone from a bridge.** Drop a stone from a $10\text{ m}$ bridge (starts at rest) — find the time to hit the water and the impact velocity. Choosing "up" as positive: $v_0 = 0$, $a = -g = -9.8\ \text{m/s}^2$, $\Delta x = -10\ \text{m}$ (it ends up *below* where it started). Formula 3: $-10 = 0 + \tfrac12(-9.8)t^2 \Rightarrow t \approx 1.43\ \text{s}$. Formula 1: $v = 0 + (-9.8)(1.43) \approx -14\ \text{m/s}$ — the negative sign, under our convention, means "downward," matching intuition.

**Worked example (upward motion) — throwing a stone up.** Throw a stone upward at $10\ \text{m/s}$; find the max height, the time to get there, and the total round-trip time. This one is worth working through its full logic, because it's a favorite source of trick questions:

- **Max height:** "max height" means the instant velocity = 0 (it's momentarily paused before falling back). Use Formula 4 with $v=0$: $0 = 10^2 + 2(-9.8)\Delta x \Rightarrow \Delta x \approx 5.1\ \text{m}$.
- **Rising time:** use Formula 1 with $v=0$: $0 = 10 + (-9.8)t \Rightarrow t \approx 1.02\ \text{s}$.
- **Falling time — using a *new* initial condition:** once at the peak, we can treat the fall back down as its own free-fall problem, starting fresh with $v_0 = 0$ at height $5.1\ \text{m}$. Solving for the time to fall that height gives *exactly* the same $1.02\ \text{s}$ as the rise. This is not a coincidence — **for a projectile that returns to its launch height, rising time always equals falling time**, a direct consequence of the motion's symmetry in time (the equations of motion look identical whether you run the clock forward from launch or backward from the peak).
- **Round-trip time:** two equally valid ways to get it — either double the rising time ($2 \times 1.02 \approx 2.04\ \text{s}$), or run Formula 3 directly for the *whole* trip with $\Delta x = 0$ (start and end at the same height): $0 = 10t - \tfrac12(9.8)t^2$, which factors to give $t=0$ (the start) or $t \approx 2.04\ \text{s}$ (the return) — same answer, two different routes.
- **Return velocity — also two ways:** either use Formula 1 from the peak (a fresh free-fall, $v_0=0$, over the falling time) or use Formula 1 directly over the whole round trip from launch. **Both give the same magnitude as the launch speed, but with the opposite sign** — the stone returns to its starting height moving exactly as fast as it left, just headed the other way.

**Common mistake #1:** flipping sign convention mid-problem — e.g., treating "up" as positive for the initial velocity, but then writing $g$ as a positive number when it should be negative in that same convention. Pick once, use everywhere.

**Common mistake #2:** for the round-trip problem, mixing up "one process, computed all at once" with "two separate processes, added together" and accidentally double-counting or skipping a piece. Both approaches are valid — just don't blend them in a way that loses track of what's been counted.

## 1.9 Projectile Motion

**The single biggest insight in this whole topic.** When something is launched at an angle (not straight up or down), gravity still only acts **vertically** — it has *zero* horizontal component. That means:

$$a_x = 0 \qquad a_y = -g$$

...and because these two directions have completely separate accelerations, **horizontal and vertical motion are entirely independent of each other.** They just happen to share the same clock (the same elapsed time $t$). This is the master key to every projectile problem: split everything into an x-problem and a y-problem, solve each one *using the ordinary four kinematics equations from 1.6*, and only let them interact through the shared variable $t$.

**Worked example 1 — throwing horizontally.** A stone is thrown horizontally at $5\ \text{m/s}$ from a $10\text{ m}$ bridge. Find the time to hit the water, the landing position, and the impact velocity.

- Horizontally: $v_x = 5\ \text{m/s}$, constant the whole flight (since $a_x=0$).
- Vertically: this is *exactly* the free-fall drop problem from 1.8 — $v_{0y}=0$, $a_y=-g$, $\Delta y = -10\ \text{m}$. Solve for $t \approx 1.43\text{ s}$ (same as before — **the horizontal throw doesn't change how long it takes to fall**, a fact that surprises people the first time they see it).
- Horizontal distance traveled: $x = v_x t = 5(1.43) \approx 7.15\ \text{m}$.
- Impact velocity: $v_y \approx -14\ \text{m/s}$ (from before), $v_x = 5\ \text{m/s}$ (unchanged) — combine as a vector: magnitude $\sqrt{5^2+14^2}$, direction $\arctan(14/5)$ below horizontal.

**The key realization to sit with:** the *time* the object spends in the air is determined **entirely by the vertical motion** — the horizontal velocity has no say in it whatsoever.

**Worked example 2 — throwing at an angle.** Kick a soccer ball at $10\ \text{m/s}$, $30°$ above horizontal — find the range (horizontal distance traveled before it returns to the same height).

- Split into components: $v_{0x} = 10\cos30°$, $v_{0y} = 10\sin30°$.
- The flying time comes entirely from the vertical motion — this is the exact same "round trip" calculation as 1.8's thrown-stone example, just using $v_{0y}$ instead of the full $10\ \text{m/s}$.
- Once you have the flight time, the range is simply $v_{0x} \times (\text{flight time})$ — a horizontal-motion calculation using a vertical-motion result as an input.

## 1.10 The General Projectile — Range and Maximum Range

**The intuition.** Rather than solving one specific angle each time, it's worth deriving the *general* formulas once, with a launch speed $v_0$ and angle $\theta$ left as symbols — both because it reveals *which* angle gives the farthest range (a favorite exam and competition question), and because it's genuinely useful to have the general shape memorized.

**Setting up:** $v_{0x} = v_0\cos\theta$, $v_{0y} = v_0\sin\theta$, $a_y = -g$.

- **Time of flight** (returning to launch height): from Formula 1 applied to the *symmetric round trip* (same idea as 1.8): $t_\text{flight} = \dfrac{2v_0\sin\theta}{g}$.
- **Max height:** from Formula 4 with $v_y=0$ at the peak: $h_\text{max} = \dfrac{(v_0\sin\theta)^2}{2g}$.
- **Range:** $R = v_{0x}\cdot t_\text{flight} = v_0\cos\theta \cdot \dfrac{2v_0\sin\theta}{g} = \dfrac{v_0^2 \cdot 2\sin\theta\cos\theta}{g} = \dfrac{v_0^2\sin(2\theta)}{g}$ (using the double-angle identity $2\sin\theta\cos\theta = \sin2\theta$).

**Finding the max-range angle.** Since $\sin(2\theta)$ maxes out at exactly $1$ when $2\theta = 90°$, i.e. $\theta = 45°$ — **for a given launch speed, 45° always gives the maximum possible range** (on level ground, no air resistance). This is worth remembering as a fact on its own, separate from being able to re-derive it.

**A subtlety worth flagging:** the letter $g$ carries through this whole derivation as a *symbol*, not a number — this is deliberate. Keeping $g$ symbolic (rather than plugging in $9.8$ early) is exactly what lets the $g$'s cancel cleanly in later steps and reveals the clean form of the answer. This habit — solve symbolically, plug in numbers last — pays off across the whole course, especially for competition-style problems where you're asked to prove a *relationship* rather than compute a *number*.

**Worked competition example (2013 F=ma style).** Two people of the same height stand at distances 20 m and 10 m from a wall. One throws a ball at 30° above horizontal; it bounces elastically off the wall (meaning: it reflects with the same speed, reversing only the horizontal component of velocity, leaving the vertical component unchanged) and is caught by the second person at the same height it was thrown. Find the initial speed $v_0$.

The way to approach this: **treat the whole path (before and after the bounce) as a single projectile motion problem**, since an elastic bounce off a vertical wall just reverses $v_x$ without touching $v_y$ — meaning the vertical motion (and therefore the total time in the air) is completely unaffected by the bounce; only the horizontal distance formula needs adjusting, since the ball's horizontal path folds back on itself at the wall. Total horizontal distance covered = distance to the wall (20 m) + distance from the wall to the second person (10 m) = 30 m — and this total plays the role of "range" $R$ in the standard projectile formula, since the vertical timing is unaffected by the bounce.

## 1.11 Graphing Projectile Motion

**The intuition.** Since x- and y-motion are independent (1.9), you get to draw *two* completely separate sets of graphs — one for each direction — and each behaves exactly like the 1-D graphs from 1.7.

- $a_x$-$t$ graph: flat at zero, the whole flight.
- $a_y$-$t$ graph: flat at $-g$, the whole flight (never changes, even at the peak!).
- $v_x$-$t$ graph: flat, nonzero (constant horizontal velocity).
- $v_y$-$t$ graph: straight line, sloping downward, crossing zero exactly at the peak of the trajectory.
- $x$-$t$ graph: straight line (constant slope, since $v_x$ is constant).
- $y$-$t$ graph: parabola (symmetric — the mirror image of the rise on the way down).

**A nice geometric fact worth proving to yourself:** if you eliminate $t$ between the $x(t)$ and $y(t)$ equations, the resulting $y$ vs. $x$ relationship is itself a parabola — which is, of course, exactly the graceful arcing shape you actually observe a thrown object trace through the air.

# UNIT 2 — Dynamics (Newton's Laws)
### *(Handouts, Classes 3–4)*

Kinematics described motion. Dynamics explains **why** it happens — what forces are present, and how they combine to produce the acceleration we described in Unit 1. This unit is the engine room of the entire course; nearly everything from here through momentum and rotation is built on the three laws below.

---

## 2.1 Force and Net Force

**The intuition.** A force is simply a push or a pull. What matters for predicting motion isn't any single force in isolation — it's the **net** force: the vector sum of every force acting on the object, all added together properly (with direction). An object can have five different forces acting on it and still not accelerate at all, if they happen to sum to zero.

**A quick word on what a "physics law" even means:** it's a relationship, verified by experiment and observation across an enormous range of situations, that reliably predicts behavior — not something "proven" the way a math theorem is proven, but something trusted because it has never yet failed a real test, and because it makes sharp, checkable predictions.

**Common mistake:** analyzing just one obvious force (like "the applied push") and predicting motion directly from it, without first checking what *else* is acting on the object (friction, gravity, normal force...) and summing everything.

---

## 2.2 Newton's First Law — The Law of Inertia

**The intuition.** If the net force on an object is zero, a moving object keeps moving in a straight line at constant speed forever, and a resting object simply stays at rest. Nothing needs to keep "supplying" motion — motion is self-sustaining in the absence of a net force. It's *changes* to motion that require an explanation.

This directly overturns a very natural, very wrong intuition most people carry around (going back to Aristotle): that objects "naturally" come to rest, and need a continuous force to keep moving. In reality, everyday objects come to rest because of friction and air resistance — real forces which are almost always present and easy to overlook. Take those away (a puck on frictionless ice, a spacecraft in deep space) and constant-velocity motion continues indefinitely with zero force required.

**Mass as the measure of inertia.** *Inertia* is the tendency of an object to resist changes in its motion. *Mass* (kg) is the precise, numerical way we measure how much inertia something has — it's one of the seven fundamental SI units (recall 1.1).

**Worked example:** A hockey puck sliding on essentially frictionless ice glides on at constant velocity with no applied force — it doesn't gradually run out of "motion" the way an everyday sliding object on carpet would. The everyday case only *looks* like it needs a continuous force because friction is quietly doing continuous negative work the whole time (see Unit 3).

**Common mistake:** assuming a moving object needs a force "in the direction it's moving" just to keep going. Velocity persists entirely on its own when net force is zero; force is only needed to *change* velocity, never to sustain it.

## 2.3 Newton's Second Law

**The intuition.** This is the equation connecting force to motion: net force causes acceleration, and the relationship is direct in one sense and inverse in another. Bigger net force → bigger acceleration (direct proportion). Bigger mass → *harder* to accelerate, i.e. *smaller* acceleration for the same force (inverse proportion) — this is exactly what "inertia" from 2.2 means quantitatively.

**The equation.** $a \propto F$ and $a \propto \dfrac{1}{m}$, combined: $a = F/m$, or in the form you'll use constantly:
$$\boxed{\vec{F}_\text{net} = m\vec{a}}$$
Here, "$\vec{F}$" in this formula always secretly means $\sum \vec{F}$, the *net* force (sometimes written $\vec{F}_\text{net}$) — never a single individual force. This is worth over-emphasizing because it's the single most common setup error in this entire unit.

**The unit.** Force is measured in Newtons (N), a derived unit: $1\ \text{N} = 1\ \text{kg}\cdot\text{m/s}^2$ — literally "the force needed to accelerate 1 kg at 1 m/s²." Force, like position and velocity, is a vector.

**Worked example:** The same net force, say 10 N, applied to a bowling ball (large mass) produces a small acceleration; applied to a golf ball (tiny mass), the same 10 N produces a huge acceleration. This is the everyday intuition for "heavier things are harder to push around," now made quantitative.

**Common mistake:** confusing force with velocity, thinking "a big force means a big velocity." A large force means a large *acceleration* — a large *rate of change* of velocity. An object can have enormous velocity with zero net force acting (constant-velocity cruising, by the First Law), or have zero velocity with a huge net force acting (the very instant it starts from rest, like a dragster launching).

## 2.4 Newton's Third Law

**The intuition.** Forces never occur alone — they come in matched pairs. If object A exerts a force on object B, then B *necessarily* exerts an equal-magnitude, opposite-direction force back on A, at the exact same instant. The two forces are equal in size ("=") but opposite in direction ("−").

**The detail that resolves almost every confusion about this law:** the two forces in an action/reaction pair act on **two different objects** — never on the same object. This is why they never cancel each other out in a single free-body diagram: a free-body diagram only ever includes forces acting *on* the one object you're analyzing, and a Third-Law partner, by definition, acts on someone else.

**Worked example:** Push on a wall with your hand. The wall pushes back on your hand, equally hard, in the opposite direction. If you draw a free-body diagram of *your hand*, you include the wall's push on your hand — but not your push on the wall (that force acts on the wall, not on your hand, so it doesn't belong in your hand's diagram at all).

**A second worked example, connecting to 2.2 and gravity:** You stand on the ground. Gravity pulls you down (Earth pulling on you); by the Third Law, you pull *up* on the Earth with an equal force. The ground also pushes up on you (the normal force, 2.9) — but that normal force's Third-Law partner is *you* pushing *down* on the ground, a completely different pair from the gravity pair. It's easy to accidentally treat gravity-on-you and normal-force-on-you as a Third-Law pair, since they happen to be equal in size when you're just standing still on flat ground — but they're not a Third-Law pair at all (they act on the same object — you — and come from different sources), they just happen to balance in that specific situation.

**Common mistake:** trying to "cancel" gravity against the normal force as if they were a Third-Law action/reaction pair. They can only ever be a Third Law pair if they act on *different* objects from *the same interaction* — gravity's partner is the pull *you* exert back on the Earth; the normal force's partner is the push *you* exert down into the ground. These are four *different* forces in total, not two.

## 2.5 Forces vs. Motion — reading the relationship between F and v

**The intuition, as a single rule of thumb:** force determines *acceleration*, not velocity — but the *direction of force relative to the current velocity* tells you exactly what kind of change is coming:

- **Force parallel to (same direction as) velocity:** speeds the object up.
- **Force antiparallel (opposite direction) to velocity:** slows the object down.
- **Force perpendicular to velocity:** changes the object's *direction* without changing its speed at all — this is the mechanism behind circular motion, which gets its own full unit later (Unit 5). Worth planting now: **a perpendicular force is precisely how you turn something without speeding it up or slowing it down.**

**Common mistake:** assuming an object's current velocity must be "explained" by some existing force in that same direction (this is really a restatement of the 2.2 mistake, but shows up constantly when reading force diagrams against a described motion).

---

## 2.6 Types of Forces

**The intuition.** It's useful to sort every force you'll encounter into two broad families:

- **Contact forces** — require the two objects to actually be touching: tension, normal force, friction, and any generic "applied" push or pull.
- **Field forces** — act at a distance, without contact: gravity, electric forces, magnetic forces.

This isn't just a vocabulary exercise — it matters practically because contact forces (like normal force and friction) are *reactive*: their size is often determined by the situation (how hard you're pushing, whether something is sliding), rather than being some fixed, pre-known value the way gravity ($mg$) is.

---

## 2.7 Gravity

**The intuition.** Near Earth's surface, every object experiences a downward pull whose strength is proportional to its mass. "Weight" *is* this gravitational force — they're the same thing, not two different concepts.

**The equation.** $F_G = mg$, always directed straight down.

**A unit note:** the pound is actually a unit of *force* (like the Newton), not mass — $1\ \text{N} = 0.225\ \text{lb}$. This is a common source of confusion in everyday (non-SI) usage, where "pounds" gets casually used as if it measured mass.

**Common mistake:** treating "mass" and "weight" as interchangeable. Mass (kg) is an intrinsic property of an object — how much matter, how much inertia. Weight (N) is a *force*, and depends on the local gravitational acceleration — your mass is identical on the Moon and on Earth, but your weight (and how hard it'd be to hold yourself up) is very different, because $g$ is different.

---

## 2.8 Tension

**The intuition.** Tension is the pulling force transmitted along a string, rope, or cable. It always points *along* the string, and always pulls (never pushes) — a rope can't shove something away.

**The idealization worth naming explicitly: the "ideal string."** Whenever a problem just says "string" or "rope" without further comment, it almost always means: massless (so it doesn't take any force to accelerate the string itself — whatever tension enters one end exits the other end unchanged) and unstretchable (fixed length, so both ends of the string are forced to move with the same speed and acceleration). These two idealizations are *why* "the tension is the same throughout the string" and "connected objects share the same acceleration" — both facts you'll use constantly starting in 2.15.

---

## 2.9 Normal Force

**The intuition.** Whenever two solid surfaces touch, each pushes back on the other, perpendicular to the contact surface — literally "normal" in the geometric sense (perpendicular). This is what prevents objects from simply passing through the floor, a table, a wall.

**The single most important thing to understand about normal force:** it is **not** some fixed, pre-known value — it's *reactive*. It equals whatever is needed, at that instant, to prevent the object from accelerating into the surface. On a flat horizontal table with nothing else pushing vertically, that happens to work out to $N = mg$ — but that's a *consequence* of the specific situation, not a universal formula.

**Worked example — why N ≠ mg in general.** Push down on a book on a table with an additional applied force of 20 N (straight down), in addition to its own weight. Now the table has to push back harder to prevent the book from accelerating into it: $N = mg + 20\ \text{N}$. Alternatively, if you *lift up* on the book with 20 N, the table needs to push less: $N = mg - 20\ \text{N}$. In both cases, $N \neq mg$ — it adjusts to whatever balance is needed.

**Common mistake:** writing $N = mg$ automatically, out of habit, without checking whether the surface is actually horizontal and whether any other vertical forces (an angled push, an incline, an elevator's acceleration) are present. This single assumption is responsible for a large fraction of dynamics mistakes across the whole unit — check it every time.

---

## 2.10 Friction — Static and Kinetic

**The intuition.** Friction always opposes *relative* sliding, or the *tendency* to slide, between two surfaces in contact. There are two regimes, and they behave quite differently:

- **Kinetic friction** — acts once the surfaces are actually sliding relative to each other. It's essentially a constant drag: $f_k = \mu_k N$, proportional to the normal force, and (perhaps surprisingly) *independent of the contact area* between the surfaces.
- **Static friction** — acts to *prevent* the start of sliding. This is the trickiest part to internalize: static friction is not a fixed value — it's exactly as large as it needs to be to keep the object from moving, matching whatever forward force is applied, **up to a maximum**: $f_{s,\max} = \mu_s N$. Once the applied force exceeds that maximum, the object breaks free and kinetic friction takes over.

**Why $\mu_k < \mu_s$ (usually):** it's harder to get something moving from a standstill than it is to keep it moving once it's already sliding — matching the everyday experience of a heavy box being hardest to budge right at the start.

**A key numerical fact:** friction coefficients ($\mu_k$, $\mu_s$) are dimensionless — they carry **no units** at all, since they're a ratio of two forces (friction ÷ normal force).

**Worked example — the four-part problem from the handout.** A horizontal force $F$ is applied to a resting 5 kg block on a table with $\mu_k = 0.4$, $\mu_s = 0.5$ ($mg = 49\ \text{N}$, so $N = 49\ \text{N}$ on flat ground; $f_{s,\max} = 0.5(49) = 24.5\ \text{N}$; $f_k = 0.4(49) = 19.6\ \text{N}$).

- **(a) $F = 0\ \text{N}$:** no applied force, so static friction is also exactly $0\ \text{N}$ (it only ever matches what's needed) — acceleration = 0.
- **(b) $F = 24\ \text{N}$:** this is *below* $f_{s,\max} = 24.5\ \text{N}$, so the block does **not** move — static friction exactly matches the push, $f_s = 24\ \text{N}$, and acceleration = 0.
- **(c) $F = 25\ \text{N}$:** this exceeds $f_{s,\max} = 24.5\ \text{N}$, so the block **breaks free** and starts sliding. Now kinetic friction applies (a *smaller*, fixed value): $f_k = 19.6\ \text{N}$. Net force $= 25 - 19.6 = 5.4\ \text{N}$, giving $a = 5.4/5 \approx 1.08\ \text{m/s}^2$.
- **(d) $F = 25.001\ \text{N}$:** essentially identical to (c) — the block is sliding, same kinetic friction, nearly the same acceleration.

**The big takeaway from parts (c) and (d):** there's a genuine **jump discontinuity** in a graph of friction (or acceleration) vs. applied force, right at $f_{s,\max}$. An infinitesimally small increase in applied force, right at that threshold, causes friction to suddenly *drop* from $f_{s,\max}$ down to the (usually smaller) kinetic value — which is also why the acceleration jumps from exactly zero to some nonzero value almost instantly, rather than ramping up smoothly.

**Common mistake:** assuming static friction is *always* at its maximum value, $\mu_s N$. It's not — static friction is reactive, matching the applied force up to that ceiling. Using $f_{s,\max}$ as if it were the actual friction force in a not-yet-moving scenario (like part (b) above) gives a wrong, larger-than-actual friction value.

## 2.11 Free-Body Diagrams & the General Problem-Solving Method

**The intuition.** A free-body diagram isolates a *single* object and draws every external force acting *on* it (never a force it exerts on something else) as an arrow from its center, in the correct direction, roughly proportional in length to its size. At this stage in the course, we treat every object as a single point (a "point mass") for translation — size and shape only start to matter once we get to rotation (Unit 7), where *where* on the object a force acts starts to matter too.

**The general method for any $F=ma$ problem:**
1. **Draw the free-body diagram** for the object in question.
2. **Sum the forces** (using the diagram) — this becomes the "left side" of Newton's Second Law.
3. **Determine the acceleration** from the described motion (is it speeding up? at rest? moving at constant velocity? constrained by a string to match another object?) — this becomes the "right side," $ma$.
4. **Set the two sides equal** (Newton's Second Law) and solve for whatever's unknown.

This four-step rhythm is worth internalizing as a habit that applies to *every* dynamics problem in this course, from a single block on a table all the way through Atwood machines and inclines.

---

## 2.12 Force at an Angle

**The intuition.** When an applied force isn't purely horizontal or vertical, split it into its horizontal and vertical components *first*, then treat each direction as its own (mostly) separate bookkeeping problem — very much like the projectile-motion split from Unit 1.9, but now for forces instead of velocities.

**Worked example.** A $100\ \text{N}$ force, applied at $30°$ above horizontal, drags a 10 kg mass across a table with $\mu_k = 0.4$. The vertical component of the applied force ($100\sin30°$) actually *reduces* the normal force (since it's pulling partially upward, helping support some of the object's weight) — so $N = mg - 100\sin30°$, **not** simply $mg$. This changes the friction ($f_k = \mu_k N$), which in turn changes the net horizontal force and the resulting acceleration. This is a very natural place for the "$N = mg$ by habit" mistake (2.9) to bite, since the force is no longer purely horizontal.

## 2.13 Inclines

**The intuition.** On a tilted surface, it's far more convenient to rotate your entire coordinate system to match the slope — one axis running *along* the incline, one *perpendicular* to it — rather than trying to force everything into ordinary horizontal/vertical axes. Gravity still points straight down as always, but *relative to the tilted axes*, it splits into two components:

- **$mg\sin\theta$** — the component *along* the incline, which is what drives the object to slide down.
- **$mg\cos\theta$** — the component *into* the incline, which is what the normal force balances (so $N = mg\cos\theta$, **not** $mg$ — this is exactly the "gotcha" flagged in 2.9).

**A geometric memory aid:** the steeper the incline (larger $\theta$), the more of gravity's pull acts *along* the slope (larger $\sin\theta$) and the less presses *into* it (smaller $\cos\theta$) — matching the intuition that a very steep slope is "mostly sliding," and a nearly flat one is "mostly just resting weight."

**Worked example.** A 10 kg block sits on a 30° incline with $\mu_k = 0.4$. Along the incline: driving force $= mg\sin30°$; opposing kinetic friction $= \mu_k(mg\cos30°)$ (using the incline's own normal force, not the flat-ground one!). The net force along the incline, divided by $m$, gives the acceleration down the slope.

**Common mistake:** using the flat-ground formula $N=mg$ on an incline. The correct incline normal force is $mg\cos\theta$, and using the wrong one cascades into a wrong friction force and a wrong final answer.

## 2.14 Two Bodies in Contact (on a horizontal table)

**The intuition.** When two blocks are pushed together (one behind the other, in direct contact, no string), they're forced to move with the *same* acceleration as long as they stay in contact — and the *only* way the front block accelerates at all is through a contact-push force from the block behind it (which, by Newton's Third Law, means the front block pushes back equally on the block behind it).

**The method:** draw a separate free-body diagram for *each* block, write Newton's Second Law for each individually (using an unknown contact force between them), and then either solve the two equations together, or — often faster — treat both blocks as a single combined system first (their interaction force cancels out entirely, since it's an internal Third-Law pair) to get the shared acceleration quickly, and only then go back to an individual block's equation to solve for the contact force itself.

**Worked example.** $m_1 = 8\ \text{kg}$, $m_2 = 2\ \text{kg}$, sitting side by side on a table with $\mu_k=0.4$, a $100\ \text{N}$ horizontal force pushing $m_1$ (which is in contact with, and pushing, $m_2$). Treating the whole 10 kg system together: net force $= 100\ \text{N} - f_{k,\text{total}}$ (friction on both blocks combined), giving a shared acceleration $a$. Then, isolating $m_2$ *alone*: the only horizontal forces on it are the contact push from $m_1$ and its own friction — solve for that contact push using $m_2 a$ and the now-known $a$.

**Common mistake:** assuming the $100\ \text{N}$ applied force acts on *both* blocks. It's applied only to $m_1$ — $m_2$ only ever "feels" $m_1$ indirectly, through the contact push between them.

## 2.15 Connected Bodies with an Ideal String — the general setup

**Recap of the two properties from 2.8 that make this whole category of problem solvable:**
- **Flexible** — tension is directed along the string.
- **Massless** — tension is the same at every point along the string (a massless piece of string can't have a net force on it, or it would need infinite acceleration).
- **Fixed length (unstretchable)** — every connected object shares the exact same magnitude of acceleration.

Whenever a problem just says "string," it's shorthand for all three of these idealizations, unless stated otherwise.

---

## 2.16 Atwood Machines (Two Masses Over a Pulley)

**The intuition.** Two hanging masses connect over a pulley by a string. Because the string is ideal (2.15), both masses share the same acceleration *magnitude* — one side goes down exactly as fast as the other side goes up — and the tension is identical throughout the string.

**The method:**
1. Draw a separate free-body diagram for each mass (each one only has two forces: its own weight, and the tension pulling it toward the pulley).
2. Pick the *actual direction of motion* (heavier mass down, lighter mass up) as positive **for both masses simultaneously** — this is the crucial step that makes the string constraint ($a_1 = a_2$) hold with consistent signs.
3. Write $F=ma$ for each mass individually, then add the two equations together — the (unknown) tension cancels out entirely, letting you solve for $a$ directly. Substitute back into either original equation to find $T$.

**Worked example.** $m_1 = 8\text{ kg}$, $m_2 = 2\text{ kg}$, connected over a light (massless) pulley. Take "$m_1$ moving down / $m_2$ moving up" as positive for both:
$$m_1g - T = m_1a \qquad T - m_2g = m_2a$$
Add: $(m_1-m_2)g = (m_1+m_2)a \Rightarrow a = \dfrac{(m_1-m_2)g}{m_1+m_2}$. Then substitute back for $T$.

**Worth comparing:** is $T$ bigger or smaller than $m_2 g$? Bigger or smaller than $m_1g$? (It always ends up *between* the two weights — makes sense, since it's simultaneously accelerating the lighter mass upward, beyond just supporting it, while only partially resisting the heavier mass's fall.)

**Common mistake:** assigning a *different* positive direction to each mass (e.g. "up is positive for $m_1$" but "down is positive for $m_2$," chosen independently). While this can be made to work if done very carefully, it's a common source of sign errors — much safer to pick the single shared direction of actual motion as positive for the whole system at once.

## 2.17 Connected Bodies with Friction and a Pulley

**The intuition.** This is the Atwood machine (2.16) combined with the incline/friction machinery (2.10, 2.13) — one mass sits on a table (subject to friction) while the other hangs freely over a pulley at the table's edge. The method is identical to 2.16, just with an extra friction term in the table-mass's equation.

**Worked example.** $m_1=8\text{ kg}$ hangs freely; $m_2=2\text{ kg}$ sits on a table with $\mu_k=0.4$, connected by a string over a pulley at the table's edge. Taking the direction of actual motion as positive throughout:
$$m_1g - T = m_1a \qquad T - f_k = m_2a, \quad \text{where } f_k = \mu_k m_2 g$$
Add to eliminate $T$: $m_1g - f_k = (m_1+m_2)a$.

**Worth exploring as a variation:** what changes if the roles are swapped — $m_2$ (2 kg) hangs, and $m_1$ (8 kg) is the one on the table? The heavier mass now provides the friction resistance, and the lighter hanging mass provides the driving force — a much smaller net driving force relative to the total mass being moved, so a noticeably smaller (possibly even negative, meaning the system doesn't move at all without checking $f_{s,\max}$ first!) acceleration.

**Common mistake:** forgetting to check, *before* assuming kinetic friction, whether the system even has enough driving force to overcome $f_{s,\max}$ in the first place (echoing 2.10) — a light hanging mass might not be able to budge a heavy mass on a high-friction table at all.

## 2.18 Static Equilibrium

**The intuition.** If an object is not accelerating (whether it's sitting still, or moving at constant velocity), the net force on it is exactly zero. Since force is a vector, "net force = 0" is secretly **two separate conditions** that both have to hold simultaneously — one for each independent direction:
$$\sum F_x = 0 \qquad \sum F_y = 0$$
(There's no $z$-direction condition needed for the 2-D problems in this course.)

**An important subtlety worth naming:** $\vec{F}_\text{net}=0$ does **not** mean the object can't be moving — it only means it isn't *accelerating*. An object cruising at constant velocity is in exactly the same equilibrium condition as one sitting perfectly still.

**Worked example.** A $2\text{ kg}$ block hangs from two ropes — one perfectly horizontal, one at $30°$ above horizontal. Break each tension into $x$- and $y$-components, then set up and solve the two equations $\sum F_x=0$, $\sum F_y=0$ *simultaneously* (they share the two unknown tensions, so you generally need both equations together, not one at a time).

**Common mistake:** solving only one of the two equations (often just the $y$-equation, since gravity is the most "obvious" force) and treating that as the complete answer. Whenever forces are at angles, you almost always need *both* equations together to pin down every unknown.

## 2.19 Measured Weight (Apparent Weight)

**The intuition.** A bathroom scale doesn't measure gravity directly — it measures the **normal force** it's exerting on you (or, by Newton's Third Law, that you're exerting on it). Ordinarily, standing still on flat ground, that normal force equals $mg$, so the scale reads your "true" weight. But the moment there's *vertical acceleration* — like in an elevator — the normal force has to change to produce that acceleration, and the scale reads something different from $mg$.

**Worked example.** A 60 kg person stands on a scale inside an elevator accelerating **upward** at $5\ \text{m/s}^2$. Taking up as positive: $N - mg = ma \Rightarrow N = m(g+a) = 60(9.8+5) = 888\ \text{N}$ — noticeably *more* than their resting weight of $588\ \text{N}$. This matches the familiar feeling of being pressed down into the floor as an elevator starts moving upward.

**The opposite case:** if the elevator instead accelerates **downward** at $10\ \text{m/s}^2$ — i.e., in true free fall — then $N = m(g-a) = m(9.8-10) \approx 0$. The scale reads essentially zero: **weightlessness**, exactly like the sensation on a fast-falling roller coaster, or an astronaut in orbit (who is, in a real sense, continuously falling — more on this in Unit 6).

**Common mistake:** confusing "weightless" with "no gravity." Gravity is still fully present and acting in both examples above — what's changed is the normal force, because the whole reference frame (elevator, roller coaster, orbiting spacecraft) is itself accelerating along with the person inside it.

## Unit 2 Summary

- **Newton's Three Laws**, in one line each: (1) no net force → no change in motion; (2) $F_\text{net}=ma$; (3) forces come in equal-and-opposite pairs acting on *different* objects.
- **Normal force and static friction are reactive** — they take whatever value is needed by the situation, up to (for static friction) a maximum. This single idea explains more dynamics "gotchas" than any other fact in the unit.
- **The four-step method** (diagram → sum forces → identify acceleration → solve) applies uniformly, from a single block to an Atwood machine to an incline with friction.
- **Ideal strings** transmit tension unchanged and force connected objects to share one common acceleration magnitude — the tool that makes multi-body problems tractable.

**The single biggest meta-mistake in this whole unit:** assuming $N=mg$ reflexively, without checking whether the surface is tilted or whether some other force has a vertical component.

# UNIT 3 — Work, Energy, and Power
### *(Handouts, Classes 5–6)*

Dynamics (Unit 2) gave you the tools to analyze motion moment-by-moment, tracking vectors and directions carefully. This unit introduces a completely different, often much faster toolkit: tracking a single *scalar* quantity — energy — as it's transferred and transformed. The trade-off is important to understand from the start: you give up detailed information about time and direction, but in exchange, many problems become dramatically simpler.

---

## 3.1 Work — Defining the Bridge Between Force and Energy

**The intuition.** Work measures the *accumulated effect of a force acting as an object moves through some distance*. It's the mechanism by which a force actually changes an object's energy — a force that acts but produces no motion (or acts perpendicular to all motion) transfers no energy at all, no matter how strong it is.

**Deriving where it comes from.** Rather than treating "work" as an arbitrary new definition, notice that it falls directly out of combining two things you already know from Units 1–2: the kinematics equation $v^2 = v_0^2 + 2a\Delta x$ (Unit 1.6) and Newton's Second Law $F_\text{net}=ma$ (Unit 2.3). Solve the kinematics equation for $a$: $a = \dfrac{v^2-v_0^2}{2\Delta x}$, and substitute into $F_\text{net}=ma$:
$$F_\text{net}\Delta x = \tfrac12 mv^2 - \tfrac12 mv_0^2$$
The left side is exactly what we'll define as **work**; the right side is a *change* in the quantity $\frac12 mv^2$, which we'll define as **kinetic energy**. In other words: work and kinetic energy weren't invented independently and then found to relate to each other — the relationship is baked into their very definitions, straight out of Newton's Second Law.

**The equation.** $W = F\Delta x$ (for a constant force, along the direction of motion). Kinetic energy: $K = \tfrac12 mv^2$. And the relationship that falls out immediately: $W_\text{net} = \Delta K$ — we'll come back to this as the **Work-Energy Theorem** in 3.6.

---

## 3.2 The Units of Work and Energy

**Checking the units make sense.** Work's left-side unit is $\text{N}\cdot\text{m}$; kinetic energy's right-side unit is $\text{kg}(\text{m/s})^2 = \text{kg}\cdot\text{m}^2/\text{s}^2$. Since $1\ \text{N} = 1\ \text{kg}\cdot\text{m/s}^2$ (from 2.3), $1\ \text{N}\cdot\text{m} = 1\ \text{kg}\cdot\text{m}^2/\text{s}^2$ — the units on both sides genuinely match, which is a nice internal-consistency check on the derivation in 3.1.

This combined unit gets its own name: the **Joule** (J). Like the Newton, it's a derived unit built from the fundamental ones, given a nickname for convenience.

---

## 3.3 Work with an Angle, and Variable Forces

**The intuition — only the component of force *along* the direction of travel counts.** If a force is applied at some angle to the direction of motion, only its component parallel to the displacement contributes to work; the perpendicular component contributes nothing at all, no matter how large.

**The equation.** $W = \vec{F}\cdot\vec{d} = F_\parallel d = Fd\cos\theta$, where $\theta$ is the angle between the force and the displacement.

**What happens when force isn't constant.** If $F$ changes as the object moves (a classic example: a spring, where $F=kx$ grows as it's stretched further), you can't just multiply $F\times d$ directly — instead, **work equals the area under a graph of $F$ versus position**. For a spring specifically, since $F=kx$ is a straight line through the origin on an $F$-$x$ graph, the "area" is the area of a triangle: $W = \frac12 kx \cdot x = \frac12 kx^2$ — a formula we'll reuse constantly starting in 3.7 and again in the Simple Harmonic Motion unit.

**Worked example — a 5 kg object, pushed by a 100 N force across 8 m on a table with $\mu_k=0.4$, starting from rest.** Compute the work done by *each* individual force acting on it:

- **Work by the applied force $F$** (at $0°$ to the motion): $W_F = Fd\cos0° = Fd$ — fully positive, since the force and displacement point the same way.
- **Work by the normal force $N$** (at $90°$ to the motion, since $N$ is vertical and the motion is horizontal): $W_N = Nd\cos90° = 0$. **The normal force never does any work on an object sliding along a surface it's perpendicular to** — worth remembering as a near-universal fact for flat-surface sliding problems.
- **Work by kinetic friction $f_k$** (at $180°$ to the motion, since friction always opposes the direction of sliding): $W_{f_k} = f_k d\cos180° = -f_k d$ — **negative**, since friction removes energy from the object's motion rather than adding to it.
- **Work by gravity $F_G$** (at $90°$ to the motion, for purely horizontal sliding): $W_{F_G}=0$, for the same reason as the normal force.

**The pattern worth internalizing:** whenever a force is perpendicular to the motion (normal force on a flat surface, gravity during purely horizontal motion, or vice versa — gravity's own direction perpendicular to *vertical* motion, etc.), it contributes **zero** work — not "small," but *exactly* zero, regardless of magnitude.

## 3.4 Work Along an Incline — Path Dependence vs. Path Independence

**The intuition, set up as a direct comparison.** Consider an object sliding down an incline of height $h$ (same height, but the incline can be made longer, shallower, or shorter, steeper — different angles $\theta$, different lengths $L = h/\sin\theta$), with friction coefficient $\mu_k$.

- **Work by gravity:** $W_G = mgh$ — and notice, this depends *only* on the height dropped, not on the incline's angle or length at all. Whether the object slides down a short, steep ramp or a long, shallow one, as long as the height dropped is the same, gravity does exactly the same amount of work. **Gravity's work doesn't depend on the path taken — only on the start and end position.**
- **Work by friction:** $W_f = -\mu_k mg\cos\theta \cdot L$ — this very much *does* depend on the specific path (the angle and length of the incline chosen), since a longer path means friction acts over a longer distance, removing more energy.

This single comparison is the seed of the entire conservative-vs-non-conservative distinction coming up in 3.7 and 3.8 — it's worth sitting with concretely before the formal definitions arrive.

## 3.5 Kinetic Energy and the Work-Energy Theorem

**Recap and restate, now as a tool rather than a derivation.** From 3.1: $W_\text{net} = \Delta K$, where $K=\frac12 mv^2$. In words: **the net work done on an object exactly equals the change in its kinetic energy.** Work is the *mechanism*; kinetic energy is the *quantity being transferred*.

**Why this is often faster than $F=ma$ directly.** The Work-Energy Theorem deals only in **scalars** — you never have to juggle vector directions the way $F=ma$ requires — and it doesn't require the force to be constant throughout the motion (a varying force just contributes a different "area under the curve" of work, but the theorem itself still holds unchanged). Whenever a problem only wants a final speed or a stopping distance, and doesn't care about the details of *how* the object got there (specifically, doesn't need velocity's *direction* mid-process, and doesn't ask about elapsed time), the Work-Energy Theorem tends to be the fastest route.

**Worked example — comparing both methods head-to-head.** A 5 kg object on a table with $\mu_k=0.4$ is moving at $10\ \text{m/s}$; how far does it slide before stopping (with no applied force, just friction decelerating it)?

- **Method 1 ($F=ma$ + kinematics):** $f_k = \mu_k mg$, so $a = -\mu_k g$ (deceleration). Use $v^2=v_0^2+2a\Delta x$ with $v=0$: solve for $\Delta x$.
- **Method 2 (Work-Energy Theorem):** $W_\text{net} = W_{f_k} = -f_k\Delta x = -\mu_k mg \Delta x$. Set this equal to $\Delta K = 0 - \tfrac12 mv_0^2$ and solve directly for $\Delta x$ — notice the mass $m$ actually cancels out of this equation entirely, which is a nice hint that the stopping distance shouldn't depend on the object's mass at all (only on $v_0$, $\mu_k$, and $g$) — a fact that's much easier to *notice* via the energy method than via the force method.

**A second worked example, showing where energy methods truly shine — measuring $\mu_k$ in the lab.** An object of mass $m$ starts from rest, slides down a frictionless incline of height $h$, then continues onto a horizontal table with friction, coming to rest after sliding a distance $d$ on the table. Find $\mu_k$.

Trying to solve this with $F=ma$ directly is painful — you'd need to separately compute the acceleration (and thus the velocity) at the bottom of the incline, and then set up a *second*, separate kinematics problem for the horizontal deceleration, carefully passing the velocity from one stage into the next. Using conservation of energy (which we formalize in 3.9) instead lets you write **one single equation** covering the *entire* process, start (at rest, height $h$) to finish (at rest, height 0, having lost energy to friction over distance $d$):
$$mgh = \mu_k mg d \quad\Rightarrow\quad \mu_k = \frac{h}{d}$$
— a strikingly clean result, and notice again that $m$ cancels entirely. **This is genuinely the practical way physicists measure a surface's coefficient of kinetic friction in a lab: drop something from a known height onto the surface and measure how far it slides before stopping.**

**Common mistake:** trying to apply the Work-Energy Theorem to a problem that explicitly asks about *time*, or that needs to track velocity's *direction* partway through (like a 2-D collision, covered in Unit 4) — the theorem has no time variable built into it at all, and treats everything as a scalar, so it simply can't answer those kinds of questions.

## 3.6 Conservative Forces and Potential Energy

**The intuition.** A force is called **conservative** if the work it does moving an object between two points depends *only* on those two points (start and end) — never on the specific path taken between them. Gravity and the spring force both have this property (as previewed in 3.4). This path-independence is precisely what allows us to associate a conservative force with a **potential energy** — a quantity that depends only on *position*, representing the energy "stored" and available to be converted back into motion.

**Defining potential energy from the conservative force's work.** $W_c = -\Delta U$ — the work done *by* a conservative force equals the *negative* of the change in the associated potential energy. (The minus sign makes intuitive sense: as an object falls and gravity does *positive* work on it, its gravitational potential energy *decreases* — the two are always trading off in opposite directions.)

- **Gravitational potential energy:** $U_g = mgh$.
- **Elastic (spring) potential energy:** $U_s = \frac12 kx^2$ (directly from the "area under the F-x graph" derivation in 3.3).

**Defining mechanical energy.** $E = K + U$ — kinetic plus potential, together.

**Common mistake:** treating the *absolute* numeric value of potential energy as if it means something on its own. It doesn't — **only differences (changes) in $U$ ever show up in the physics.** You are always completely free to choose your own reference point for "zero height" (or "zero potential energy") — the specific number you assign to $U$ at any one point is arbitrary, but $\Delta U$ between any two points is not.

---

## 3.7 Non-Conservative Forces

**The intuition.** A force is **non-conservative** if the work it does *does* depend on the path taken — more distance covered against that force means more energy lost, even between the same two starting and ending points. In this course, the only non-conservative force you'll deal with is **kinetic friction** — and its work always goes toward heat (thermal energy), which is why it's sometimes described as "leaking" mechanical energy out of the system rather than just moving it between kinetic and potential forms.

**Worked example (revisiting 3.4).** Two paths between the same start and end points on an incline, one steep-and-short, one shallow-and-long, both with the same $\mu_k$: gravity does the *same* work on both (conservative — same height drop), but friction does *more negative* work on the longer path (non-conservative — more distance means more energy lost to heat).

---

## 3.8 The Work-Energy Theorem, Reorganized — Three Categories, Three Relations

**The big-picture summary the handout emphasizes:** there are three categories of work (by conservative forces, by non-conservative forces, and their sum — the net), three categories of energy (kinetic, potential, and their sum — mechanical), and three relations tying them together:

$$W_\text{net} = \Delta K \qquad W_c = -\Delta U \qquad W_\text{nc} = \Delta E$$

**Where the third relation comes from.** Since $W_\text{net} = W_c + W_\text{nc}$ and $W_\text{net}=\Delta K$, substituting $W_c=-\Delta U$: $-\Delta U + W_\text{nc} = \Delta K \Rightarrow W_\text{nc} = \Delta K + \Delta U = \Delta(K+U) = \Delta E$.

**Why the third relation ($W_\text{nc}=\Delta E$) is usually the most useful one in practice:** it lets you put the *entire* effect of friction (or any other non-conservative force) on one side of an equation, and the *entire* change in the system's total mechanical energy on the other — without ever needing to separately track $W_c$ or worry about the path-dependence of conservative forces at all (since they're already baked into $\Delta U$).

---

## 3.9 Power

**The intuition.** Power measures *how quickly* energy is transferred or work is done — it's a rate, not an amount. Doing the exact same amount of work in half the time requires twice the power.

**The equation.** $P = W/t$ (unit: $\text{J/s}$, nicknamed the **Watt**, W). Since $W=F\Delta x$ and average velocity $v=\Delta x/t$, substituting gives an equally important alternate form: $P = Fv$, or more generally with an angle, $P = \vec{F}\cdot\vec{v} = Fv\cos\theta$.

**Worked example — why cars slow down on hills.** A car's engine has a maximum power output it can sustain. Climbing a hill requires a larger net driving force than cruising on flat ground (to overcome the added component of gravity along the incline, echoing 2.13). Since $P=Fv$ and $P$ is capped by the engine, a **larger required force directly forces a smaller sustainable velocity** — which is exactly why cars visibly slow down on steep hills even at full throttle, without anything being "wrong" with the engine.

**Other named units worth knowing (a nice tangent on unit "nicknames"):** force, work, and power are each derived units that got their own convenient names (Newton, Joule, Watt) rather than being written out in raw kg/m/s combinations every time. A few more nicknamed energy units you'll encounter outside this course: $1$ Calorie $=4.18$ J (the heat-energy unit from chemistry); $1$ Kilocalorie $= 1000$ Calories (the "Calorie," capital C, used on food labels); $1$ Horsepower $=746$ W (the traditional power unit for engines).

**Common mistake:** confusing power with either force or total energy. A small motor and a large motor can do *identical total work* (deliver the same energy) — the only difference is *how long it takes* the smaller, lower-power motor to get there.

## 3.10 Conservation of Mechanical Energy

**The intuition — the single most powerful idea in this unit.** If the *only* forces doing work on a system are conservative (i.e., $W_\text{nc}=0$ — no friction, no other energy-dissipating forces), then total mechanical energy never changes at all: it simply trades back and forth between kinetic and potential forms, but the *sum* stays exactly constant.

**The equation.** From $W_\text{nc}=\Delta E$ (3.8), setting $W_\text{nc}=0$ gives $\Delta E = 0$, i.e.:
$$E_1 = E_2 \qquad \text{i.e.} \qquad K_1+U_1 = K_2+U_2$$
The subscripts "1" and "2" can refer to two different *locations*, or two different *moments in time* — whichever framing fits the specific problem.

**Worked example — free fall, revisited from an energy point of view.** Drop an object from height $h$; find its speed just before hitting the ground, using energy instead of kinematics. $K_1+U_1 = K_2+U_2 \Rightarrow 0+mgh = \tfrac12 mv^2+0 \Rightarrow v=\sqrt{2gh}$ — matching exactly what Formula 4 from Unit 1 would give, but arrived at without ever separately solving for time or acceleration.

**Worked example — a frictionless incline gives the *same* final speed as free fall from the same height.** This is worth sitting with as a genuine "aha": a ball rolled down *any* frictionless ramp shape, or dropped straight down, from the same starting height $h$, reaches the bottom at *exactly* the same speed $\sqrt{2gh}$ — because kinetic energy at the bottom only depends on how much potential energy was converted, which only depends on the height dropped, never on the specific path.

**Worked example — the roller coaster.** Release a cart from rest at height $h$ at the top of a frictionless track that dips down and later rises to a smaller height. Find the speed at the bottom, and at any other point of known height, entirely through $E_1=E_2$, without ever needing to know (or use $F=ma$ on) the shape of the track itself — which would be essentially impossible to do directly, since the track's curvature means the net force (and thus the acceleration) is constantly changing direction and magnitude throughout the ride.

- **Choosing your reference height matters for bookkeeping, not for the physics.** If you choose the bottom of the track as "height zero," then $U=0$ there and all the initial PE converts directly to KE. If instead you choose the *top* of a subsequent hill (say, height $2R$ for a loop of radius $R$) as your reference, the numbers you write down look different, but the final computed *speeds* come out identical — because, again, only $\Delta U$ ever matters physically (3.6).

**A related worked example — does launch direction matter for final speed?** Throw a $0.5\ \text{kg}$ stone at $10\ \text{m/s}$ from a height of $2\ \text{m}$, in four different directions: straight down, purely horizontal, at $45°$, and straight up. Which one hits the ground with the greatest speed?

Using energy conservation immediately reveals the answer *without* doing four separate kinematics problems: $E_1=E_2 \Rightarrow \tfrac12 mv_0^2 + mgh = \tfrac12 mv_f^2$. Notice that **the initial speed $v_0$ enters this equation only through $\frac12 mv_0^2$ — as a magnitude, with no dependence on direction at all.** So the final speed at ground level is **exactly the same in all four cases** — a genuinely surprising result if you only think in terms of vector velocities, but an immediate, almost trivial consequence of energy conservation, since kinetic energy only cares about *speed*, never direction.

**Common mistake:** forgetting that conservation of energy in this simple form ($E_1=E_2$) only applies when $W_\text{nc}=0$. The moment friction (or another non-conservative force) is present, you must go back to the more general $W_\text{nc}=\Delta E$ and actually include that term — a very common error is to keep using the frictionless shortcut on a problem that explicitly has friction somewhere in it.

## 3.11 The General Conservation of Energy

**The intuition — zooming out beyond just mechanics.** Everything above has been about *mechanical* energy specifically (kinetic + potential), and the condition for it to be conserved (no non-conservative forces doing work). But there's a much broader, essentially universal principle sitting behind this, one you'll meet again in chemistry, and again if you ever study general physics further:

> "Energy cannot be created or destroyed; it may be transformed from one form into another, but the total amount of energy never changes." — Albert Einstein

The friction "losses" we've been tracking as $W_\text{nc}$ aren't actually energy vanishing from the universe — they're energy *converting* into heat (thermal energy of the surfaces involved), which is just as real and just as conserved as kinetic or potential energy; we simply don't track it as part of *mechanical* energy in this course.

# UNIT 4 — Impulse and Momentum
### *(Handouts, Classes 7–8)*

Work and energy (Unit 3) tracked the accumulated effect of force over **distance**. This unit tracks the accumulated effect of force over **time** instead — a parallel toolkit, especially powerful for anything involving sudden interactions: collisions, explosions, impacts.

---

## 4.1 Impulse

**The intuition.** Impulse answers: "how much did this force push, and for how long?" It's the direct time-based cousin of work (which is the distance-based accumulation of force).

**Deriving it (mirroring 3.1's derivation, but from a different kinematics equation).** Start from Formula 1, $v=v_0+at$, and $F_\text{net}=ma$. Solve Formula 1 for $a$: $a = \dfrac{v-v_0}{t}$, substitute into $F_\text{net}=ma$:
$$F_\text{net}t = mv-mv_0$$
The left side is what we'll define as **impulse**; the right side is a *change* in the quantity $mv$, which we'll define as **momentum** (next section). Exactly as with work and kinetic energy, impulse and momentum aren't independently invented — they fall directly out of Newton's Second Law, just via a different one of the four kinematics equations.

**The equation.** $J = F\Delta t$ for a constant force. If the force varies with time, impulse is the **area under an $F$-$t$ graph** (mirroring the "area under $F$-$x$" rule for variable-force work in 3.3).

---

## 4.2 Momentum

**The intuition.** Momentum is a way of capturing "how much motion" an object has, blending both how heavy it is *and* how fast it's going into a single vector quantity. Colloquially: "how hard would this be to stop?"

**The equation.** $p = mv$ (vector, unit $\text{kg}\cdot\text{m/s}$).

**Worked comparison.** A slow-moving loaded truck and a fast-moving small car can easily have the *same* momentum — momentum weighs mass and velocity together, so neither one alone tells the whole story. This is part of why momentum (rather than velocity alone, or mass alone) turns out to be exactly the right quantity for describing collisions in this unit.

---

## 4.3 The Impulse-Momentum Theorem

**The relation.** $J = F\Delta t = \Delta p = mv - mv_0$. A quick unit check confirms the derivation: $\text{kg}\cdot\text{m/s} = \text{N}\cdot\text{s}$ — both sides genuinely match.

**Worked example 1 — a baseball being struck.** A $0.145\ \text{kg}$ baseball is struck by a bat for $1.5\ \text{ms}$, with the force rising and falling in a roughly triangular shape peaking at $F_\max = 18{,}000\ \text{N}$.
- **(a) Impulse delivered:** since the force-time graph is a triangle, the area (impulse) is $\frac12 \times F_\max \times \Delta t = \frac12(18{,}000)(0.0015) = 13.5\ \text{N}\cdot\text{s}$.
- **(b) Average force:** by definition, $F_\text{avg} = J/\Delta t = 13.5/0.0015 = 9000\ \text{N}$ — notice this is exactly *half* of $F_\max$, which makes sense for a symmetric triangular force profile.
- **(c) Speed after being struck:** $\Delta p = J \Rightarrow v = J/m$ (assuming it started at rest, or you'd use $\Delta p = m(v-v_0)$ more generally).

**Worked example 2 — the stuntman, and why airbags save lives.** An 80 kg stuntman jumps from a 45 m window.

- **(a) Speed at ground level:** this is a free-fall problem straight out of Unit 1 (or Unit 3's energy shortcut): $v=\sqrt{2gh}$.
- **(b) Landing directly on the ground, impact time $10\ \text{ms}$:** the *same* momentum change $\Delta p = mv$ (from the speed found in part (a), to zero) has to occur — but crammed into an extremely short time, so the required force $F = \Delta p / \Delta t$ is enormous.
- **(c) Landing on an air-filled target instead, impact time $1.5\ \text{s}$ (150 times longer):** the momentum change $\Delta p$ is **exactly the same** as in part (b) — same starting and ending velocity — but spread across a much longer time, so the required force is **150 times smaller**.

**The core physical insight, worth sitting with:** the *momentum change is identical* in both landing scenarios (b) and (c) — what airbags, crumple zones, and padded surfaces actually do is not reduce how much your momentum has to change, but **stretch out the time over which that change happens**, which (from $F=\Delta p/\Delta t$) directly and proportionally reduces the peak force involved. This is precisely why cars have crumple zones and why airbags deploy over a fraction of a second rather than instantly: the goal is never to avoid the momentum change (that's unavoidable, physically, given the collision), only to control how quickly it happens.

**Common mistake:** assuming a "gentler," lower-force collision must involve a smaller momentum change. It doesn't — for the same start and end velocity, $\Delta p$ is fixed; what differs between a violent collision and a gentle one is purely the *time* over which that fixed $\Delta p$ occurs, and therefore the *force* required.

## 4.4 Conservation of Momentum

**The intuition.** If the net *external* force on a system is zero, its total momentum never changes. The key word is **external** — internal forces (like the forces two colliding objects exert on each other) always come in Newton's-Third-Law pairs (Unit 2.4), and when you sum the momentum of the *whole system together*, those internal pairs always cancel out exactly, leaving only external forces with any power to change the total.

**The equation.** If $F_{\text{net, ext}}=0$, then $\Delta p_\text{total} = 0$ — i.e., $\sum p_\text{before} = \sum p_\text{after}$.

**Why this is fundamentally different from conservation of energy.** Conservation of *energy* (Unit 3.10) is a statement that can apply to a *single* object (an object's own mechanical energy is conserved if no non-conservative force acts on *it*). Conservation of *momentum*, by contrast, is inherently a statement about a **system of two or more interacting objects** — momentum conservation doesn't even make sense to apply to a lone, non-interacting object (its momentum trivially "conserves" only in the sense that nothing is happening to it at all).

**Where this shows up in collisions.** In a typical tabletop collision problem, gravity and the normal force are present (external, vertical) but balance each other exactly, giving zero *net* external force — and the actual **collision force between the two objects is internal**, extremely large but very brief, and (crucially) cancels out of the system-total momentum by Newton's Third Law. This is exactly why we can (and almost always should) apply conservation of momentum to collisions, even though a large, complicated, hard-to-measure force is present *during* the collision itself.

**Common mistake #1:** applying conservation of momentum to a single, isolated object as if it were conservation of energy — remember, it's fundamentally a *system* statement.

**Common mistake #2:** forgetting momentum is a *vector*. In two-dimensional collisions, you must conserve momentum **separately in each direction** ($x$ and $y$) — not as some combined magnitude.

---

## 4.5 Collisions — Setting Up the General Approach

**The method.** Write down the total momentum equation for the system:
$$\sum p_i = \sum p_f \qquad \text{i.e.} \qquad m_1v_1+m_2v_2 = m_1v_1'+m_2v_2'$$
Both velocities and momenta are vectors here — apply this relation to each independent direction as needed.

**The catch: one equation, two unknowns.** In a typical two-object collision, both final velocities ($v_1'$ and $v_2'$) are unknown — that's *two* unknowns, but conservation of momentum alone only gives you *one* equation. You need **one additional piece of information**, almost always coming from what's known (or assumed) about the *kinetic energy* of the collision — which is exactly what the three collision categories below provide.

---

## 4.6 Three Categories of Collisions

**Sorted by what happens to kinetic energy:**

- **Elastic** — kinetic energy is fully conserved (nothing is lost to heat, sound, or deformation). This is an idealization, but a very good approximation for things like billiard balls or hard steel spheres.
- **Inelastic** — some kinetic energy is lost (converted to heat, sound, deformation).
- **Perfectly inelastic** — the two objects stick together after colliding, moving with one shared final velocity. This is the case where the **most** kinetic energy possible (for that specific momentum) is lost.

Note: **momentum is conserved in *all* three categories** — it's only kinetic energy whose fate differs between them.

---

## 4.7 Perfectly Inelastic Collisions

**The intuition.** "Stick together" gives you the second equation you needed for free: $v_1' = v_2' = v'$ (one shared final velocity), which turns the momentum equation into something solvable with just one unknown.

**Worked example.** Two objects $m_1$, $m_2$ collide and stick together, forming a single combined mass $m_1+m_2$.
$$m_1v_1+m_2v_2 = (m_1+m_2)v' \quad\Rightarrow\quad v' = \frac{m_1v_1+m_2v_2}{m_1+m_2}$$
To find the energy lost, compute $K_i$ (before, using the original two velocities separately) and $K_f = \frac12(m_1+m_2)v'^2$ (after), and take the difference — you should always find $K_i > K_f$ for a perfectly inelastic collision (energy is lost, never gained).

## 4.8 Elastic Collisions

**The intuition.** Here, the *second* equation comes from kinetic energy being fully conserved:
$$\tfrac12 m_1v_1^2 + \tfrac12 m_2v_2^2 = \tfrac12 m_1v_1'^2 + \tfrac12 m_2v_2'^2$$
Combined with conservation of momentum, you now have two equations and two unknowns ($v_1'$, $v_2'$) — solvable, though it requires a bit of algebra (a quadratic ends up appearing, since kinetic energy involves squared velocities). It's worth simply remembering the **general elastic-collision result formulas** for competition-style speed, rather than re-deriving them from scratch every time — though understanding *where* they come from (the two conservation equations above) is what lets you rebuild them if you forget the exact form.

**A memorable special case worth knowing on its own: equal masses.** If $m_1=m_2$ in a head-on elastic collision, the two objects **completely exchange velocities** — the moving object stops, and the previously-stationary object moves off with exactly the first object's original velocity (for the case where one starts at rest). This is a genuinely useful shortcut to recognize instantly on sight, rather than grinding through the general algebra, whenever a problem specifically involves equal masses.

**Common mistake:** forgetting that the "quadratic formula gives two solutions" issue (mentioned as a competition-exam reminder in the handout) often has one *trivial* solution (like $v_1'=v_1$, meaning "no collision happened") mixed in with the physically meaningful one — always check which root actually describes the collision that took place, and sanity-check the resulting directions of motion make physical sense.

## 4.9 Inelastic Collisions (General Case — Some, But Not All, Energy Lost)

**The intuition.** Between the two extremes (elastic: no energy lost; perfectly inelastic: sticking together, maximum energy lost for that momentum) lies the general "inelastic" case, where you're typically told directly *how much* kinetic energy is lost (often as a fraction, like "half the kinetic energy is lost") — and that stated fraction becomes your second equation, replacing either the "energy fully conserved" or "objects stick together" conditions.

**Worked example.** A mass $m$ moving at speed $v$ collides with an identical stationary mass $m$; half the kinetic energy is lost in the collision. Find the velocities after. Here, momentum conservation gives one equation ($mv = mv_1'+mv_2'$), and "$K_f = \frac12 K_i$" gives the second — together enough to solve for both final velocities (though, since there are still two unknowns and this is a less constrained scenario than elastic or perfectly-inelastic, you may need an additional assumption, like both moving in the same final direction, to fully pin down a unique answer).

---

## 4.10 Collisions That Aren't Head-On (Two Dimensions)

**The intuition.** Once objects don't collide along a single line, you need the full vector nature of momentum (flagged as a common mistake back in 4.4): **conserve momentum separately in each direction**, $x$ and $y$.

**A powerful simplification worth remembering:** if, along *one* particular direction, the impulse is zero throughout the whole process, then the system's momentum *in that direction alone* is conserved on its own — even if momentum isn't simply conserved as a single combined vector equation the way it is in 1-D.

**Worked example.** A mass $m$ moving east at $v_1$ collides with a mass $M$ moving north at $v_2$; they stick together (perfectly inelastic). Set up **two separate momentum equations** — one for the east-west direction, one for the north-south direction — and solve for the two components of the final combined velocity separately; then combine those two components (via the Pythagorean theorem and an arctangent) to get the final speed and direction as a single vector answer. The energy lost is found the usual way: $K_i$ (computed separately for each object beforehand) minus $K_f$ (using the combined mass and the combined final speed).

## 4.11 Explosions

**The intuition.** An explosion is, in a sense, the exact time-reverse of a perfectly inelastic collision: a single object (mass $m$) separates into two (or more) fragments, with the **total momentum before the explosion equal to the total momentum of all fragments after** — conservation of momentum applies here just as much as it does to a collision, since the explosive forces involved (however violent) are entirely internal to the system.

**A key difference from ordinary collisions: kinetic energy actually *increases*** during an explosion — chemical energy stored in the explosive is converted *into* kinetic energy, the reverse of the "lost to heat" story in inelastic collisions.

**Worked example (setup).** A large mass $m$, initially at rest, separates into two smaller masses $m_1$ and $m_2$. Since the initial momentum was zero, conservation of momentum requires the two fragments to fly off with **equal and opposite momenta**: $m_1v_1' = -m_2v_2'$ — meaning the lighter fragment moves away *faster* than the heavier one, in exactly the ratio that keeps their momenta balanced.

---

## 4.12 Center of Mass

**The intuition.** The center of mass is the single "balance point" of a system — the one point that behaves, for the purposes of predicting overall motion, as if the *entire* mass of the system were concentrated there, and as if every *external* force acted directly on it.

**The equations.** $x_\text{cm} = \dfrac{m_1x_1+m_2x_2}{m_1+m_2}$ (and the same form for $y$); this generalizes directly to any number of masses. The center of mass's own velocity: $v_\text{cm} = \dfrac{m_1v_1+m_2v_2}{m_1+m_2}$.

**The single most important fact about center of mass, and exactly *why* momentum conservation works the way it does:** internal forces — a collision, an explosion, anything happening entirely within the system — can **never** change $v_\text{cm}$. Only a genuinely *external* force can accelerate the center of mass. This isn't a coincidence sitting alongside conservation of momentum — it's really the *same fact*, restated: since $p_\text{total} = (m_1+m_2)v_\text{cm}$, saying "total momentum is conserved" and saying "$v_\text{cm}$ doesn't change" are two ways of expressing the identical idea.

**Worked example, tying it together.** In a two-object collision (elastic, inelastic, or perfectly inelastic — doesn't matter which), the center of mass of the *system* continues moving in a perfectly straight line at perfectly constant velocity throughout the entire collision, completely unaffected by whatever violent, complicated things are happening between the two individual objects during the impact.

## 4.13 Worked Example — Applying Center of Mass to a Projectile Explosion

**Setup (from the handout).** A shell is fired at $v_0=20\ \text{m/s}$, at $60°$ above horizontal. At the very peak of its trajectory, it explodes into two equal-mass fragments. One fragment has zero velocity immediately after the explosion and falls straight down. How far from the launch point does the *other* fragment land, assuming level ground and no air resistance?

**The key insight that makes this tractable: track the center of mass, not the individual fragments, for the parts of the motion where that's easier.** Before the explosion — and, crucially, this *continues to be true after* the explosion too, since the explosion is purely an internal event — the center of mass of the (now two-fragment) system continues to follow **exactly the trajectory the original, unexploded shell would have followed** — a perfectly ordinary projectile parabola, landing at the ordinary projectile range (Unit 1.10) for a $20\ \text{m/s}$, $60°$ launch.

Since one fragment (call it Fragment A) simply drops straight down from the peak — landing directly below the peak's horizontal position, at exactly half the ordinary range (since the peak of a projectile's arc sits at the horizontal midpoint of its full range) — and since the center of mass of the two *equal*-mass fragments must land at the position dictated by the *undisturbed* projectile range (call this $R$), you can use the **definition of center of mass position** to back out where Fragment B lands: with equal masses, the center of mass sits exactly halfway between the two fragments' landing points, so
$$R = \frac{x_A + x_B}{2} \quad\Rightarrow\quad x_B = 2R - x_A$$
This lets you solve the whole problem almost entirely using tools from Unit 1 (ordinary projectile range) plus this unit's center-of-mass idea — without ever needing to work out the explosion's internal forces or the individual fragment velocities directly.

## Unit 4 Summary

- **Impulse and momentum** are the time-based parallel to work and kinetic energy — both pairs fall directly out of Newton's Second Law, just via different kinematics equations.
- **The Impulse-Momentum Theorem** ($J=\Delta p$) is the key to understanding safety technology: airbags and crumple zones don't reduce the *momentum change* in a crash (that's fixed by the physics), they extend the *time* over which it happens, directly reducing the *force* involved.
- **Conservation of momentum** applies to *systems*, not single objects, and only because internal (Third-Law) forces always cancel when summed across the whole system.
- **The three collision categories** (elastic, inelastic, perfectly inelastic) differ only in what happens to kinetic energy — momentum conservation applies to all three equally.
- **Center of mass** ties directly back to momentum conservation: internal forces (collisions, explosions) can never move the center of mass — only external forces can.

**The single biggest meta-mistake in this unit:** trying to solve a two-unknown collision problem with momentum conservation alone. You always need a second equation — from the type of collision (elastic, perfectly inelastic) or from explicitly given information about energy lost.

# UNIT 5 — Uniform Circular Motion
### *(Handout, Class 9)*

Every unit so far has dealt with straight-line motion, or motion split into independent straight-line pieces (like projectiles). This unit introduces genuinely new territory: motion around a circle. The central theme, worth holding onto throughout: **circular motion is fundamentally the motion of an angle**, and almost everything here is a direct, deliberate parallel to Unit 1's kinematics — just re-expressed in terms of rotation instead of straight-line displacement.

---

## 5.1 The Four Angular Quantities

**The intuition.** Just as Unit 1 needed position, displacement, velocity, and acceleration to describe straight-line motion, circular motion needs its own four analogous quantities — but built from **angle** instead of position:

- **Angular position, $\theta$** — where you are, expressed as an angle from some reference direction.
- **Angular displacement, $\Delta\theta$** — how much the angle changed.
- **Angular velocity, $\omega = \Delta\theta/\Delta t$** — how fast the angle is changing.
- **Angular acceleration, $\alpha = \Delta\omega/\Delta t$** — how fast the angular velocity itself is changing.

**A crucial unit note:** $\theta$ must be measured in **radians** for these formulas to work correctly (not degrees) — radians are the "natural" angular unit because they directly relate arc length to radius ($s=r\theta$), which is exactly what lets the linear/angular formulas in 5.4 come out clean.

**These are all vector quantities**, but circular motion (confined to a flat plane) only really needs a simplified sense of direction: **counterclockwise (CCW) or clockwise (CW)** — usually just tracked with a sign (CCW positive, by the most common convention) rather than full vector notation.

---

## 5.2 Angular Kinematics — the Direct Parallel to Unit 1

**The intuition.** For constant angular acceleration $\alpha$, you get **the exact same four formulas as Unit 1.6**, just with every linear quantity relabeled as its angular counterpart:

$$x\to\theta \qquad v\to\omega \qquad a\to\alpha$$

$$\omega = \omega_0+\alpha t \qquad \Delta\theta = \tfrac12(\omega_0+\omega)t \qquad \Delta\theta = \omega_0t+\tfrac12\alpha t^2 \qquad \omega^2=\omega_0^2+2\alpha\Delta\theta$$

**Why this parallel works so cleanly — and why it's worth trusting rather than re-deriving from scratch:** the *mathematical derivation* in Unit 1.6 never actually used anything specific to straight-line motion — it only used the definitions of "rate of change" applied twice in a row. Since $\omega=\Delta\theta/\Delta t$ and $\alpha=\Delta\omega/\Delta t$ have the *exact same structure* as $v=\Delta x/\Delta t$ and $a=\Delta v/\Delta t$, the entire derivation carries over line-for-line. This is a genuinely efficient way to learn this section: rather than treating it as sixteen new facts to memorize, treat it as four facts you already know, wearing a different set of labels.

**Worked example.** A wheel starts from rest and spins up with a constant angular acceleration of $2\ \text{rad/s}^2$ for $3\ \text{s}$. Using the direct parallel to the very first worked kinematics example back in Unit 1.6: $\omega = 0+2(3)=6\ \text{rad/s}$, and $\Delta\theta = 0+\tfrac12(2)(3)^2=9\ \text{rad}$ — solved by exactly the same reasoning ("what am I not given?") as any Unit 1 kinematics problem.

**Common mistake:** mixing linear and angular quantities together without the proper conversion factor (covered next, in 5.4) — for instance, writing $v=\omega$ directly, which is dimensionally wrong ($v$ is in m/s, $\omega$ is in rad/s — they're only related through a multiplication by radius $r$).

## 5.3 Period, Frequency, and Angular Frequency — Describing Repetition

**The intuition.** Since circular motion repeats itself over and over, it's useful to describe *how often* it repeats, in addition to the instant-by-instant angular quantities from 5.1–5.2.

- **Period, $T$** — how much time one full revolution takes.
- **Frequency, $f = 1/T$** — how many revolutions happen per unit time; unit: $\text{s}^{-1}$, nicknamed the **Hertz (Hz)**.
- **Angular frequency, $\omega = 2\pi f = 2\pi/T$.**

**A subtlety worth flagging carefully, since the handout explicitly calls it out:** this "$\omega$" (angular frequency) is technically a **scalar** — it just describes *how fast* the repeating cycle happens, with no inherent direction — whereas the "$\omega$" from 5.1 (angular *velocity*) is a vector-like quantity with a rotational sense (CCW/CW). **For uniform circular motion specifically, these two numerically coincide** (the constant rotation rate *is* both the angular velocity's magnitude and the angular frequency), which is why they're written with the same symbol and often used interchangeably in this context — but conceptually, it's worth keeping the distinction in mind, especially once you reach Simple Harmonic Motion (Unit 8), where "angular frequency" reappears in a context that has nothing to do with actual physical rotation.

---

## 5.4 Tangential Velocity and Acceleration — Connecting Linear and Angular

**The intuition.** A point moving around a circle of radius $r$ has both an angular description (how its angle is changing) and an ordinary linear description (how far it's actually physically moving, in meters, at that instant). Geometry connects the two directly, via the arc-length relationship $\Delta s=r\Delta\theta$ (this is, not coincidentally, exactly why radians were the natural angular unit to use in 5.1 — with radians, this relationship has no extra conversion constant cluttering it).

**The equations.** Dividing $\Delta s=r\Delta\theta$ by $\Delta t$: $v_t = r\omega$. Differentiating (or repeating the same argument one level up): $a_t = r\alpha$.

**Why "tangential":** at any instant, this velocity/acceleration points along the direction the object is *actually, instantaneously* traveling — tangent to the circle at that point — as opposed to the *radial* direction (5.5), which points toward or away from the center.

---

## 5.5 Radial (Centripetal) Acceleration

**The intuition — the key realization from Unit 1.4, now made fully precise.** Recall from Unit 1.4 that *any* change in velocity counts as acceleration — including a change in direction alone, even at constant speed. Uniform circular motion is the clean, canonical example of exactly this: constant speed, but *constantly* changing direction, therefore *constantly* accelerating.

**Deriving the direction and size of this acceleration.** Consider velocity at two nearby instants, $\vec{v}_1$ and $\vec{v}_2$, both with the same magnitude (uniform speed) but slightly different directions (having rotated along with the object). The difference vector $\Delta\vec{v}=\vec{v}_2-\vec{v}_1$ — worked out geometrically — points **toward the center of the circle**, not along the direction of travel at all. That's the origin of the name "centripetal" (center-seeking).

**The equations.** Two equivalent forms, depending on which quantities you have on hand:
$$a_c = \frac{v^2}{r} \qquad\qquad a_c = r\omega^2$$
(the second form follows immediately from substituting $v=r\omega$, 5.4, into the first).

**Putting it together — the general acceleration of a particle in circular motion.** In general (not necessarily *uniform*), a particle moving in a circle can have **both** a tangential component (speeding up or slowing down along the circle, from 5.4) **and** a radial/centripetal component (constantly redirecting it toward the center, from this section) simultaneously:
$$\vec{a} = \vec{a}_t + \vec{a}_c$$
These two components are always perpendicular to each other (tangent vs. radial), so their magnitudes combine via the Pythagorean theorem if you ever need a single combined magnitude.

---

## 5.6 What "Uniform" Actually Means

**The intuition, precisely stated.** "Uniform circular motion" means the **speed** $|\vec v|$ is constant — nothing more, nothing less. Digging into what that implies: constant speed means $\omega$ is constant too (5.4's relation $v=r\omega$, with fixed $r$), which in turn means $\alpha=0$ and therefore $a_t=0$ (5.4's relation $a_t=r\alpha$).

**But — and this is the crucial catch worth sitting with:** even though $a_t=0$, the **centripetal** acceleration $a_c=v^2/r$ is emphatically **not** zero — it's constant in *magnitude*, but constantly changing in *direction* (always pointing toward whatever the current center-ward direction is, which rotates along with the object).

**Two genuinely good conceptual-check questions worth answering carefully, straight from the handout:**

- **Is velocity uniform, for uniform circular motion?** No — velocity is a vector, and its *direction* is constantly changing, even though its magnitude (speed) is constant. "Uniform" here refers only to the speed.
- **Is acceleration uniform?** No, for the same reason — the centripetal acceleration has constant *magnitude* but constantly rotating *direction*, so as a vector, it is not constant/uniform at all, even in uniform circular motion.

## 5.7 Centripetal Force

**The intuition — the single most important conceptual point in this whole unit.** Centripetal force is **not** a new, distinct type of force alongside tension, gravity, friction, and normal force. It's simply the **name for whatever net force happens to be pointing toward the center** and causing the circular path — a role that different actual forces fill in different situations, never an independent physical push of its own.

**The equation.** $F_c = F_\text{net} = ma_c$ — literally just Newton's Second Law, applied specifically to the centripetal (radial) direction.

**Worked examples of what "plays the role" of centripetal force, situation by situation:**
- A ball on a string, swung in a horizontal circle — the **string tension** is the centripetal force.
- A car turning on a flat road — **friction** between tires and road is the centripetal force.
- A satellite orbiting a planet — **gravity** is the centripetal force (a full unit's worth of consequences from this single fact, coming up in Unit 6).

**Is $F_c$ uniform (for uniform circular motion)?** By the exact same logic as 5.6: no — its *magnitude* ($mv^2/r$) is constant, but its *direction* constantly rotates to keep pointing at the (moving, in the sense of "current") center.

**How much work does centripetal force do?** This is a lovely, quick application of Unit 3's work definition ($W=Fd\cos\theta$): the centripetal force is, by definition, always **perpendicular** to the velocity (it points radially, while velocity points tangentially) — so $\cos\theta=\cos90°=0$, and **centripetal force does zero work, always**, regardless of how large it is or how long it acts. This is exactly *why* uniform circular motion maintains constant speed forever: with zero work being done, kinetic energy (which only depends on speed) never changes.

**Common mistake:** drawing "centripetal force" as an *extra*, separate arrow in a free-body diagram, in addition to the tension/gravity/friction/normal-force arrows that are actually present. It's not a separate force to add — it's simply the *label* for whichever of those already-drawn forces (or combination of them) happens to sum to something pointing toward the center.

## 5.8 Worked Numerical Example — Same Object, Three Different Setups

**Setup:** $m=0.5\ \text{kg}$, $r=0.2\ \text{m}$, $v=2\ \text{m/s}$ (constant speed circular motion in each case — only the *orientation* of the circle, and therefore which forces are involved, changes).

**(a) Mass on a table, string providing the centripetal force (horizontal circle):** the string tension points horizontally, directly toward the center, and *is* the entire centripetal force: $T = mv^2/r$.

**(b) Mass on a vertical circle, at the very top:** now there are **two** forces with components toward the center (which, at the top, points straight down): both gravity and the string tension point downward (toward the center of a vertical circle, when the mass is at the top). Together, they must sum to the required centripetal force: $T + mg = mv^2/r \Rightarrow T = mv^2/r - mg$.

**(c) Mass on a vertical circle, at the very bottom:** here, gravity still points down, but the center (and thus the required centripetal direction) is now *up* — so gravity actually points **away** from the center at this point, and the tension has to work against gravity *and* still supply the full centripetal requirement: $T - mg = mv^2/r \Rightarrow T = mv^2/r + mg$.

**The pattern worth noticing:** the *same* object, same speed, same radius, requires a **different** tension depending purely on where gravity happens to be pointing relative to the center at that instant — largest at the bottom (gravity fighting against the needed force), smallest at the top (gravity helping supply it).

## 5.9 Critical Speeds for Vertical Circular Motion

**The intuition.** Part (b) above (mass on a string, at the top of a vertical circle) hides an important subtlety: what if the speed is slow enough that the *required* centripetal force is actually *less* than gravity alone would provide? A string can only **pull**, never push — so if gravity alone would already over-provide the centripetal force needed, the string would have to push outward to compensate, which it physically cannot do. This gives rise to a **minimum speed** requirement, below which the string goes slack and the object can no longer maintain the circular path (it falls away from the circle instead).

**Finding $v_\min$ for a string, at the top:** the critical (borderline) case is exactly when $T=0$ — the string is on the verge of going slack, contributing nothing, and gravity is doing the *entire* job of supplying the centripetal force:
$$mg = \frac{mv_\min^2}{r} \quad\Rightarrow\quad v_\min = \sqrt{gr}$$
**How to judge that this is really the *minimum*, not the maximum:** if $v$ were any *smaller* than this, the required centripetal force ($mv^2/r$) would be *less* than what gravity alone supplies — meaning gravity would be "overqualified," pulling the mass in faster than the circular path calls for, so the mass falls inward, off the circular path, rather than maintaining it. If $v$ is *larger*, the string simply supplies the extra needed tension ($T=mv^2/r - mg > 0$) — no problem there. So the critical value is a floor, not a ceiling, for the string case.

**A closely related but inverted scenario: a car driving over the crest of a hill.** Here, the normal force from the road can only **push up** (away from the road surface), never pull down — the mirror-image constraint of the string case. At the top of a hill (treating the hilltop as part of a large circle), gravity points down (toward the center of that circle) and the normal force points up (away from the center) — so $mg - N = mv^2/r$. As speed increases, $N$ must *decrease* to keep supplying more of the centripetal requirement from gravity's share... until $N$ hits zero, at which point the car is on the verge of **losing contact with the road entirely** (becoming momentarily airborne) if it goes even faster:
$$mg = \frac{mv_\max^2}{r} \quad\Rightarrow\quad v_\max = \sqrt{gr}$$
**How to judge this is a *maximum*:** if the car goes any faster than this, gravity alone can't supply enough centripetal force to keep the car curving down and along with the hill's surface — the car flies off the road's surface (briefly airborne) instead. So here the critical value is a ceiling, not a floor — the opposite role from the string case, even though the formula looks identical.

**The common thread worth extracting:** both critical-speed problems come from the exact same technique — identify which force in the situation is a "cannot-push" or "cannot-pull" constraint (a string can't push; a road surface can't pull a car down onto it), set that force's contribution to exactly zero, and solve what's left.

## 5.10 Critical Speeds for Horizontal Turning

**The intuition.** Turning a corner (rather than going over a vertical loop) still requires a centripetal force — now pointing horizontally, toward the center of the curve. Two classic setups, each with its own "critical speed" logic:

**(a) A curved, flat road relying on friction.** Here, **static** friction (2.10) supplies the centripetal force — static, not kinetic, because the tire's contact patch with the road isn't sliding (ideally); it's just being pushed sideways by the road as the car turns. Since static friction can be *anything* up to a maximum, there's a **maximum speed** for taking the curve without sliding out:
$$f_{s,\max} = \mu_s N = \mu_s mg \quad (\text{on flat ground}) \quad = \quad \frac{mv_\max^2}{r} \quad\Rightarrow\quad v_\max = \sqrt{\mu_s gr}$$
**How to judge this is a maximum:** exceeding it means the required centripetal force would exceed the *maximum* static friction can supply — at that point, the tires begin to slide (switching to the smaller kinetic friction value), and the car cannot maintain the curve; it skids outward, off the intended circular path.

**(b) A banked (tilted) curve, frictionless.** Rather than relying on friction at all, race tracks and highway curves are often tilted (banked) so that the **normal force itself** has a horizontal component pointing toward the center of the curve — allowing the turn to be taken even with zero friction, at exactly one specific "ideal" speed for a given bank angle and radius. Splitting the normal force into components (much like the incline analysis in Unit 2.13, but now used for a *horizontal* centripetal purpose rather than resisting sliding down a slope): the vertical component of $N$ balances gravity, and the horizontal component of $N$ supplies the entire centripetal force:
$$N\cos\theta = mg \qquad N\sin\theta = \frac{mv^2}{r}$$
Dividing these two equations (a classic trick that eliminates both $N$ and $m$ at once): $\tan\theta = \dfrac{v^2}{gr}$.

**Is this a minimum, maximum, or an exact ideal value?** For the frictionless banked case, there's no "up to a maximum" force involved (unlike static friction) — it's a single **exact** required speed for that specific bank angle to work with zero friction needed at all. (In practice, real banked curves also have some friction available, which is what allows a *range* of safe speeds around this ideal value, rather than only one exact speed working.)

**The unifying observation, worth remembering as the takeaway for the whole unit:** every one of these "critical speed" problems — vertical string, hilltop car, flat curve with friction, banked curve — is solved with the exact same three-step process: (1) identify what's supplying the centripetal force in this specific setup, (2) write $F_\text{net, centripetal} = mv^2/r$ using only the forces actually present, and (3) figure out whether the answer represents a floor, a ceiling, or an exact ideal value, based on which force (if any) has a hard physical limit (can't push, can't pull, can't exceed a max).

## Unit 5 Summary

- **Circular motion is fundamentally the motion of an angle** — every one of Unit 1's linear kinematics tools has a direct angular parallel ($x\to\theta$, $v\to\omega$, $a\to\alpha$).
- **"Uniform" means constant speed only** — velocity and acceleration, as vectors, are still constantly changing direction throughout.
- **Centripetal force is a role, not a new force** — always identify which already-familiar force (tension, gravity, friction, normal force) is actually filling that role in a given problem.
- **Centripetal force always does zero work**, since it's always perpendicular to velocity — this is exactly why speed stays constant in uniform circular motion.
- **"Critical speed" problems** all come from the same technique: find which force has a hard physical limit (can't push, can't pull, maxes out), set it to that limiting value, and solve.

**The single biggest meta-mistake in this unit:** treating centripetal force as an extra force to add to a free-body diagram, rather than recognizing it as the *label* for the net inward force already supplied by tension, gravity, friction, or normal force.

# UNIT 6 — Newton's Law of Gravitation
### *(Handout, Class 10)*

Back in Unit 2, gravity showed up as a simple, fixed downward force: $F_G=mg$. This unit reveals that formula as just a **local approximation** to something much more general — a universal force that acts between *any* two masses, anywhere, and whose consequences stretch all the way out to satellite orbits, Kepler's laws, and the tides.

---

## 6.1 Newton's Law of Universal Gravitation

**The intuition.** Every object with mass attracts every other object with mass — not just "the Earth pulling on things," but a completely general, mutual attraction between *any* two masses anywhere. The strength of this pull grows with the masses involved, and shrinks rapidly (as an inverse-square) with the distance between them.

**The equation.**
$$F_G = \frac{Gm_1m_2}{r^2}, \qquad G = 6.67\times10^{-11}\ \text{N}\cdot\text{m}^2/\text{kg}^2$$

**Why you've never noticed this force between everyday objects.** $G$ is an extremely small number — so although gravity technically acts between, say, you and your desk, the resulting force is so tiny it's utterly undetectable in practice. **In practice, this force only becomes noticeable when at least one of the two masses is astronomically large** (a planet, a moon, a star) — which is why, throughout this course, "gravity" almost always specifically means "the pull of the Earth (or another large body) on something," rather than the pull between two small objects.

**Common mistake:** using $r$ as the distance between the *surfaces* of two objects. The formula always uses the distance between their **centers** — for a person standing on Earth's surface, $r$ is the distance from Earth's center to the person, which is essentially Earth's radius, *not* some tiny surface-to-surface gap.

---

## 6.2 The Relation Between $F_G=Gm_1m_2/r^2$ and $F=mg$

**The intuition.** These aren't two competing formulas — $F=mg$ is simply what Newton's general Law of Gravitation reduces to, specifically at Earth's surface. Setting them equal for an object of mass $m$ at Earth's surface (distance $R$ from Earth's center, with Earth's mass $M$):
$$mg = \frac{GmM}{R^2} \quad\Rightarrow\quad g = \frac{GM}{R^2}$$
Plugging in Earth's actual values ($R = 6.37\times10^6\ \text{m}$, $M=5.98\times10^{24}\ \text{kg}$) reproduces the familiar $g\approx9.8\ \text{m/s}^2$ — or, thought of slightly differently, $g$ can be understood as Earth's local **gravitational field strength**, $9.8\ \text{N/kg}$.

**Why $F=mg$ is only valid "near the surface."** The formula $g=GM/r^2$ genuinely depends on $r$ — the distance from Earth's center. But since Earth's radius ($\sim6370\ \text{km}$) is enormous compared to any height $h$ you might climb, drop something from, or throw something to, the *change* in $r$ (and therefore in $g$) over any everyday height is negligible — $g$ stays essentially constant, and $F=mg$ works beautifully. But **once $h$ becomes comparable to Earth's radius itself** (satellites, orbits, anything discussed for the rest of this unit), $g$ can no longer be treated as constant, and you must go back to the full inverse-square law.

**Common mistake:** using $F=mg$ (with the surface value of $g=9.8\ \text{m/s}^2$) for a satellite or any object at a genuinely large altitude. The surface value of $g$ only applies *near* the surface — for anything meaningfully far away, you need the full $F_G=GmM/r^2$, using the actual (larger) $r$.

## 6.3 Gravitational Potential Energy — the General Form

**The intuition.** Unit 3 introduced $U_g=mgh$ for gravitational potential energy — but exactly like $F=mg$ itself, this is only a *local* approximation, valid when $g$ can be treated as roughly constant (near Earth's surface). For situations spanning large distances (satellites, escape velocity), you need the *general* gravitational potential energy, derived by integrating the general force law over distance (the same "area under a force-vs-position graph" idea from Unit 3.3 and 3.6, just now for a force that varies as $1/r^2$ rather than linearly like a spring):
$$U_G = -\frac{GmM}{r}$$

**The sign is worth pausing on.** This potential energy is *always negative*, and it *increases* (becomes less negative, approaching zero) as $r$ grows — meaning objects have their *lowest* (most negative) potential energy when very close together, and their potential energy approaches (but never quite reaches) zero as they're separated infinitely far apart. This is a different convention from $U_g=mgh$ (which is zero at your chosen reference height and grows positively from there) — and that difference is entirely a matter of *where each formula chooses to call potential energy "zero."* $U_g=mgh$ sets zero at your reference height; $U_G=-GmM/r$ sets zero at infinite separation.

**Common mistake — explicitly flagged in the handout: never mix the two formulas together in the same problem.** Since they use *different* zero-reference conventions, you cannot use $U_g=mgh$ for one part of a calculation and $U_G=-GmM/r$ for another part of the *same* calculation and expect the numbers to combine sensibly — pick whichever is appropriate for the length scales involved (near-surface: $mgh$; large-distance/orbital: $-GmM/r$) and stay consistent throughout that problem.

---

## 6.4 Applying to Satellites — Orbital Velocity

**The intuition — the entire idea of an orbit in one sentence.** A satellite "in orbit" isn't defied by gravity — it's *entirely* controlled by gravity, which supplies exactly the centripetal force (Unit 5.7) needed to keep it moving in a circle rather than flying off in a straight line (as Newton's First Law would otherwise dictate).

**The equation, derived by setting gravity equal to the required centripetal force:**
$$\frac{GmM}{r^2} = \frac{mv^2}{r} \quad\Rightarrow\quad v = \sqrt{\frac{GM}{r}}$$
Notice: the satellite's *own* mass $m$ cancels out completely — orbital speed depends only on the central body's mass $M$ and the orbital radius $r$, **never** on how heavy the orbiting object itself is. A tiny cubesat and a massive space station, at the same altitude, orbit at exactly the same speed.

**Worked example (qualitative).** Since $v\propto 1/\sqrt{r}$, satellites in *lower* orbits must travel *faster* to maintain their circular path (gravity is stronger there, requiring more centripetal speed to match), while satellites farther out orbit more slowly — consistent with observing that a low-Earth-orbit satellite circles the globe roughly every 90 minutes, while the Moon (much farther away) takes about 27 days.

## 6.5 The Energy of a Satellite

**The intuition.** Now that both kinetic energy ($K=\tfrac12mv^2$) and the general gravitational potential energy ($U_G=-GmM/r$) are on the table, it's worth combining them for an orbiting satellite and seeing what falls out.

**Deriving it, step by step.** From 6.4, $v^2=GM/r$, so:
$$K = \tfrac12 mv^2 = \frac{GmM}{2r} = \tfrac12|U_G|$$
— a clean, memorable relationship: **for a circular orbit, kinetic energy is always exactly half the magnitude of the potential energy.** Adding them for the total mechanical energy:
$$E = K+U = \frac{GmM}{2r} - \frac{GmM}{r} = -\frac{GmM}{2r} = -K$$
**The total orbital energy is negative**, and (worth noting) is exactly the *negative* of the kinetic energy alone — another clean, memorable relationship specific to circular orbits.

**What the negative sign physically means.** A negative total energy means the satellite is **gravitationally bound** — it doesn't have enough energy to escape to infinity (where $U_G\to0$) and coast away forever; it's "trapped" in orbit, the same way a ball rolled into a valley doesn't have enough energy to climb back out over the rim.

---

## 6.6 Escape Velocity

**The intuition.** "Escape velocity" is the minimum launch speed needed for an object to break completely free of a planet's gravity and never fall back — reaching $r\to\infty$ with, at minimum, zero velocity left over (the borderline case: just barely escaping, arriving at infinity moving infinitesimally slowly).

**Deriving it.** Set total energy at launch (surface, radius $R$) equal to total energy at infinity (where both $K$ and $U_G$ are zero, in the borderline "just barely escapes" case):
$$\tfrac12 mv_\text{esc}^2 - \frac{GmM}{R} = 0 \quad\Rightarrow\quad v_\text{esc} = \sqrt{\frac{2GM}{R}}$$
For Earth, this works out to about $v_\text{esc}\approx1.12\times10^4\ \text{m/s}$ (roughly 11.2 km/s, or about 25,000 mph) — a genuinely enormous speed, which is a large part of why leaving Earth's gravity well is so difficult and fuel-intensive.

**Connecting back to 6.5:** notice escape velocity corresponds *exactly* to the condition "total energy = 0" — the precise dividing line between a bound orbit (negative total energy, like 6.5) and successfully escaping to infinity (zero or positive total energy).

## 6.7 Geosynchronous Satellites and Kepler's Third Law

**The intuition.** A "geosynchronous" satellite orbits with the **same period** as Earth's own rotation (so it appears to stay above the same point on Earth's surface, as seen from the ground) — an enormously useful property for communication and weather satellites.

**Kepler's Third Law**, which we'll properly introduce in 6.10, gives the relationship needed to find such an orbit's radius:
$$\frac{R^3}{T^2} = \text{constant}$$
For a geosynchronous satellite, $T=86{,}400\ \text{s}$ (one full day) — solving gives an orbital radius of about $r=4.23\times10^7\ \text{m}$, or a height above Earth's surface of about $h=3.59\times10^7\ \text{m}$ (roughly 22,000 miles) — this is, in fact, exactly why GPS and communications satellites sit at this specific, very particular altitude.

**A fun aside worth working through yourself: how fast is the ground itself moving, just from Earth's own rotation?** Earth completes one full rotation ($2\pi$ radians) in one day, giving an angular velocity of $\omega\approx0.0000727\ \text{rad/s}$. At the equator (radius roughly Earth's radius, $6.37\times10^6\ \text{m}$), the tangential speed (5.4's $v=r\omega$) works out to about $463\ \text{m/s}$ — meaning that just by sitting still on the equator, you're already moving over 1000 mph relative to Earth's center, and you travel roughly 40,000 km (Earth's circumference) every single day, entirely "for free," just by staying put.

**Some real period values worth having a sense of scale for, mentioned in the handout:** the synodic month (time between successive full moons) is about 29.53 days; the tropical year (time for Earth's seasons to repeat) is about 365.242 days — these numbers being *not quite* whole numbers is precisely why calendars need leap years and occasional adjustments.

## 6.8 Angular Momentum of a Satellite, and Kepler's Second Law

**The intuition.** A satellite orbiting a much larger central body experiences essentially **zero torque** about that central body (gravity, acting directly along the line connecting the two centers, has no leverage to twist the orbit — more on why this specific geometry produces zero torque once you reach Unit 7.2). With no torque acting, the satellite's **angular momentum is conserved** throughout its orbit (a concept we'll properly build up in Unit 7.6) — and this single fact, applied to an *elliptical* orbit (rather than a perfect circle), is exactly what produces Kepler's Second Law.

**Kepler's Second Law, stated:** a line drawn from the central body to the orbiting body sweeps out **equal areas in equal times** — meaning the satellite moves *faster* when it's closer to the central body (near part of the ellipse) and *slower* when it's farther away, in precisely the ratio that keeps angular momentum (and swept-out area) constant throughout.

---

## 6.9 Kepler's Three Laws — the Full Picture

**1st Law — The Ellipse.** Planetary (and satellite) orbits are ellipses, not perfect circles, with the central body located at one of the two foci of the ellipse (not at the center). This is the actual solution to the full gravitational equations of motion — a genuinely nontrivial mathematical result, though the *particular* shape (which specific ellipse) depends on the initial conditions of the orbit.

**2nd Law — Equal Areas in Equal Times.** As derived in 6.8, directly from angular momentum conservation.

**3rd Law — $R^3/T^2 = \text{constant}$.** As used already in 6.7 — worth noting that this relationship reduces to the circular-orbit velocity formula from 6.4 as a special case (a circle is, after all, just a special case of an ellipse), with gravity again playing the role of the centripetal force.

---

## 6.10 The Tides

**The intuition.** Tides are mainly caused by the **Moon's** gravitational pull on Earth (the Sun contributes too, but less, despite its far greater mass, because tidal effects depend on the *difference* in gravitational pull across Earth's diameter, which falls off faster with distance than the pull itself does — making the much-closer Moon the dominant contributor). Because Earth rotates once per day while the tidal bulge stays roughly aligned with the Moon, most coastal locations experience **two high tides per day** as Earth rotates underneath that bulge.

**Spring tides and neap tides.** When the Sun and Moon are aligned (new moon or full moon), their tidal effects add together, producing unusually large ("spring") tides. About a week later, when the Sun and Moon are at right angles relative to Earth, their effects partially cancel, producing unusually small ("neap") tides.

- **Newton's Law of Gravitation** ($F_G=Gm_1m_2/r^2$) is the universal version of the "gravity" you met in Unit 2 — $F=mg$ is just its local approximation near a planet's surface.
- **Gravitational potential energy** has a *different* zero-reference convention ($U_G=-GmM/r$, zero at infinite separation) than the near-surface $U_g=mgh$ — never mix the two in one calculation.
- **Orbital motion is just circular motion (Unit 5) where gravity supplies the centripetal force** — leading to orbital speed, orbital energy, and escape velocity, all following directly from setting gravity equal to $mv^2/r$ or from energy bookkeeping.
- **Kepler's Three Laws** describe the general (elliptical) case, with the Second Law following directly from angular momentum conservation, and the Third Law generalizing the circular-orbit speed relationship.

**The single biggest meta-mistake in this unit:** using the near-surface formulas ($F=mg$, $U_g=mgh$, with a fixed $g=9.8\ \text{m/s}^2$) for problems involving large distances or altitudes, where the full inverse-square law is required instead.

# UNIT 7 — Torque and Rotational Motion
### *(Handouts, Classes 11–12)*

Up through Unit 6, every object has been treated as a single point — a "point mass" with no size or shape. This unit finally lets objects have actual **extent**: they can spin, and *where* on the object a force is applied starts to matter, not just how big that force is. Much of this unit will feel familiar, because it's built as a deliberate, careful parallel to Units 1–4 — just with rotation standing in for translation throughout.

---

## 7.1 Rotational Kinematics — Building on Unit 5

**The intuition.** The angular kinematics you already met in Unit 5 (angular position, velocity, acceleration, and the four constant-$\alpha$ formulas) apply completely unchanged here — rotation is rotation, whether you're calling it "circular motion of a point" or "spinning of an extended object." The only genuinely new idea in this unit is **rotational dynamics**: what *causes* an object to spin up or slow down, and how an object's shape and mass distribution affects that.

**A key setup detail for this unit:** objects here rotate either about their **center of mass** or about some other **fixed axis** — and the question this unit answers is: given a force applied somewhere on the object, how much does that force actually contribute to spinning it?

---

## 7.2 Torque

**The intuition.** Torque is the rotational analog of force — it's what actually causes angular acceleration. But unlike ordinary force, torque cares about *two* things at once: how big the force is, *and* how far from the pivot (and at what angle) it's applied. A small force applied far from the pivot, at a good angle, can out-twist a much larger force applied close to the pivot or along a bad angle.

**The equation.** $\tau = F\cdot d_\perp = F_\perp \cdot d = Fd\sin\theta$, where $d$ is the distance from the pivot to where the force is applied (sometimes called the "lever arm" or "moment arm"), and $\theta$ is the angle between the force and that lever arm. Unit: $\text{N}\cdot\text{m}$ — notice this happens to be dimensionally identical to the Joule (from work, Unit 3.2), but **torque is never given the nickname "Joule"** — it's a fundamentally different kind of quantity (work is a scalar measuring energy transfer; torque is more like a vector, measuring twisting tendency), and keeping the units unnamed here is a deliberate reminder not to confuse the two.

**Worked example (directly from the handout) — the effect of both distance and angle.** Calculate the torque for a $10\ \text{N}$ force applied at $1\ \text{m}$, $0.5\ \text{m}$, and $0\ \text{m}$ from the pivot (holding the angle at $90°$ throughout, i.e., $\sin\theta=1$): torques of $10\ \text{N}\cdot\text{m}$, $5\ \text{N}\cdot\text{m}$, and $0\ \text{N}\cdot\text{m}$ respectively — **directly proportional to distance**, and vanishing entirely once the force is applied *at* the pivot itself (makes sense: pushing exactly at the hinge can't twist anything around that same hinge).

Now hold the distance fixed at $1\ \text{m}$ and vary the angle: $90°$, $0°$, and $30°$, giving torques of $10\ \text{N}\cdot\text{m}$, $0\ \text{N}\cdot\text{m}$, and $5\ \text{N}\cdot\text{m}$ respectively. **The $0°$ case is worth sitting with carefully:** a force applied *exactly along* the lever arm (pointing straight toward or away from the pivot) produces **zero** torque, no matter how large the force is — pushing or pulling a door handle straight *into* the hinge line doesn't twist the door at all; only the component perpendicular to the lever arm has any twisting effect.

**Everyday intuition for the $\sin\theta$ dependence:** pushing on a door far from its hinges, perfectly perpendicular to the door's face, is maximally effective (small force, easy to open). Pushing near the hinges, or pushing at a bad angle (nearly along the door's own plane, toward or away from the hinge line), takes far more force to achieve the same effect, or accomplishes almost nothing at all.

**Direction, simplified.** Like angular velocity in Unit 5, torque is treated here with a simplified sense of direction: CCW or CW, rather than full 3-D vector notation.

**Common mistake:** forgetting the $\sin\theta$ factor entirely and just multiplying $F\times d$ regardless of angle. Pushing exactly along a wrench's handle, however hard, produces zero torque — only the perpendicular component of the force contributes.

## 7.3 The Law for Rotation — Deriving $\tau_\text{net}=I\alpha$

**The intuition.** Just as $F=ma$ governs translational motion, there should be a rotational equivalent connecting torque to angular acceleration. Rather than treating this as a brand-new law to memorize, it's worth deriving it directly from $F=ma$ — which reveals *exactly* what "moment of inertia" (the rotational analog of mass) really means, rather than leaving it as an arbitrary new definition.

**The derivation.** Start with Newton's Second Law for a single point mass at distance $r$ from a rotation axis: $F_\text{net}=ma$. Multiply both sides by $r$ (a purely algebraic step, but a meaningful one):
$$F_\text{net}\cdot r = mar$$
The left side, $F_\text{net}\cdot r$, is exactly the torque (7.2, for a force perpendicular to the radius, the common simplifying case). On the right side, substitute the tangential-acceleration relation from Unit 5.4, $a=a_t=r\alpha$:
$$\tau_\text{net} = m(r\alpha)r = mr^2\alpha$$
Defining $I=mr^2$ (for now, a single point mass) as the **moment of inertia**, this becomes:
$$\boxed{\tau_\text{net} = I\alpha}$$
**This is Newton's Second Law for rotation** — and notice, it isn't a new independent law of nature at all; it falls directly out of the ordinary $F=ma$, exactly the way work-energy and impulse-momentum did in earlier units.

---

## 7.4 Moment of Inertia — For Point Masses

**The intuition, directly from the derivation above.** Moment of inertia, $I$, is the rotational analog of mass — it measures how much an object "resists" a change in its spin. But unlike ordinary mass, $I=mr^2$ depends not just on *how much* mass there is, but critically on *how far that mass sits from the axis of rotation* — squared, no less, which makes the distance-dependence quite strong.

**For a system of several point masses, sum over all of them:** $I=\sum m_ir_i^2$ — each mass contributing according to its own distance from whichever axis you've chosen (and different axis choices genuinely give different $I$ values for the *same* physical object, as the next example shows vividly).

**Worked example (directly from the handout).** Two point masses, $5.0\ \text{kg}$ and $7.0\ \text{kg}$, mounted $4.0\ \text{m}$ apart on a light (massless) rod. Find the moment of inertia about four different axes:

- **(a) Axis through the $5.0\ \text{kg}$ mass:** that mass contributes nothing ($r=0$); only the $7.0\ \text{kg}$ mass (at $r=4.0\ \text{m}$) contributes: $I = 7.0(4.0)^2 = 112\ \text{kg}\cdot\text{m}^2$.
- **(b) Axis through the $7.0\ \text{kg}$ mass:** symmetric to (a): $I = 5.0(4.0)^2 = 80\ \text{kg}\cdot\text{m}^2$.
- **(c) Axis halfway between the two masses:** now both masses are $2.0\ \text{m}$ from the axis: $I = 5.0(2.0)^2 + 7.0(2.0)^2 = 20+28 = 48\ \text{kg}\cdot\text{m}^2$ — notice this is **smaller** than either (a) or (b), which makes sense: the mass is now, on average, closer to the axis than in either single-mass-through-the-axis case.
- **(d) Axis $0.50\ \text{m}$ to the left of the $5.0\ \text{kg}$ mass:** now the $5.0\ \text{kg}$ mass is at $r=0.5\ \text{m}$, and the $7.0\ \text{kg}$ mass is at $r=4.5\ \text{m}$: $I = 5.0(0.5)^2+7.0(4.5)^2 = 1.25+141.75=143\ \text{kg}\cdot\text{m}^2$ — the **largest** of all four cases, since the axis is now farther from *both* masses on average than any of the other choices.

**The key lesson from this one example, worth remembering as the takeaway for the whole section:** the moment of inertia of a fixed physical arrangement of masses is **not** a single fixed number — it depends entirely on *which axis* you choose to rotate it about, and can vary quite dramatically (here, from 48 to 143 $\text{kg}\cdot\text{m}^2$) just by moving the axis around.

## 7.5 Moment of Inertia for Common Shapes

**The intuition.** Real objects aren't just a couple of point masses — they're continuous distributions of mass. Computing $I=\sum m_ir_i^2$ for a continuous shape technically requires calculus (essentially summing over infinitely many infinitesimal mass pieces), but the *results*, for a handful of common, symmetric shapes, are worth simply knowing:

| Shape (rotating about its central axis, unless noted) | Moment of Inertia |
|---|---|
| Ring / hoop (all mass at radius $R$) | $I=mR^2$ |
| Rod, about its center | $I=\tfrac{1}{12}mL^2$ |
| Rod, about one end | $I=\tfrac13mL^2$ |
| Disk or solid cylinder | $I=\tfrac12mR^2$ |
| Solid sphere | $I=\tfrac25mR^2$ |
| Thin spherical shell | $I=\tfrac23mR^2$ |

**A pattern worth noticing, to help remember (rather than purely memorize) this table:** the coefficient in front of $mR^2$ (or $mL^2$) essentially measures **how spread out the mass is, on average, relative to the axis**. A ring has *all* its mass at the maximum possible distance $R$ from the axis, giving it the largest coefficient (a full $1$) among the round shapes here. A solid sphere has mass distributed all the way from the center (contributing nothing) out to the surface (contributing the most), averaging out to a smaller coefficient ($\frac25$) than a disk (where all the mass is spread across a flat 2-D disk rather than filling a 3-D volume, giving $\frac12$). A rod about its *end* has a larger moment of inertia than about its *center*, since rotating about the end puts, on average, more of the rod's length (and thus more of its mass) farther from the axis.

**Worked comparison, foreshadowing 7.9:** a solid sphere ($I=\frac25mR^2$) has a *smaller* moment of inertia than a thin spherical shell of the same mass and radius ($I=\frac23mR^2$) — the shell has all its mass concentrated at the maximum radius, while the solid sphere has mass distributed throughout its interior, much of it closer to the axis. This difference has a very real, observable consequence when the two shapes are raced rolling down an incline (see 7.9).

## 7.6 Two Masses Connected Over a Pulley *With Mass*

**The intuition — revisiting the Atwood machine (Unit 2.16), now with a crucial new detail.** Previously, we always assumed an "ideal," massless pulley — one that changes the *direction* of the string's tension without requiring any force (or torque) to spin it up itself. If the pulley genuinely has mass $M$ and radius $R$, it now has its **own** moment of inertia, and needs a net *torque* to angularly accelerate along with the string moving over it.

**Why this changes the tensions on the two sides.** For a massless pulley, tension is identical throughout the whole string (Unit 2.15) — but for a pulley with real mass, the *string* tension on each side must actually be slightly **different**, because the *difference* between the two tensions is exactly what supplies the net torque needed to spin the pulley itself up ($\tau_\text{net} = (T_1-T_2)R = I_\text{pulley}\alpha$).

**Comparing to the massless-pulley case (Unit 2.16):** as the pulley's own mass $M\to0$ (or its moment of inertia becomes negligible relative to the hanging masses), this correction vanishes and $T_1\to T_2$, recovering exactly the simpler Unit 2.16 result. **The condition for the two tensions to be (nearly) equal is precisely that the pulley's own moment of inertia is small enough to be negligible** compared to what's needed to accelerate the whole system — which is exactly the hidden assumption behind calling a pulley "ideal" or "light" in earlier units.

## 7.7 Static Equilibrium — Now Including Torque

**The intuition, extending Unit 2.18.** An object at rest needs more than just zero *net force* — it also needs zero *net torque*, or it would start rotating even while its center of mass stays put. This gives a **third** equilibrium condition, on top of the two ($\sum F_x=0$, $\sum F_y=0$) from Unit 2.18:
$$\sum F_x=0 \qquad \sum F_y=0 \qquad \sum\tau=0$$
**A subtlety worth flagging, straight from the handout:** a force $\vec{F}$ and its exact opposite $-\vec{F}$, applied at *different* points on an object, can satisfy $\sum\vec{F}=0$ (they cancel as a net force) while still producing a net *torque* (if they're not applied along the same line) — this pair can rotate the object even though it won't translate at all. Conversely, two *equal, same-direction* forces $\vec{F}$, $\vec{F}$ can produce zero net torque about a well-chosen axis while still very much causing translation. **Force balance and torque balance are genuinely independent conditions — satisfying one doesn't automatically satisfy the other.**

**Worked example — the seesaw (directly from the handout).** A uniform board serves as a seesaw pivoting at point P. Child A (30 kg) sits 2.5 m from the pivot; where must Child B (25 kg) sit, distance $x$ from the pivot on the other side, to balance?

Since neither child is moving, apply $\sum\tau=0$ about the pivot P (a convenient choice of axis, since it eliminates the unknown pivot support force from the torque equation entirely — that force acts *at* the pivot, so its lever arm, and therefore its torque about that point, is automatically zero):
$$m_Ag(2.5) = m_Bg(x) \quad\Rightarrow\quad x = \frac{m_A(2.5)}{m_B} = \frac{30(2.5)}{25} = 3.0\ \text{m}$$
Then, to find the **supporting force from the pivot**, apply the *force* equilibrium condition ($\sum F_y=0$) separately: the pivot's upward support force must balance the combined downward weights of both children plus the board itself: $F_N = m_Ag+m_Bg+Mg$ (where $M$ is the board's own mass, assumed to act at its center, which is directly over the pivot in this symmetric setup — contributing zero torque about P, but still contributing to the *force* balance).

**The key technique worth extracting:** choosing the pivot point itself as the axis for the torque equation is almost always the smartest move in equilibrium problems, since it makes the (often unknown) pivot force's own torque vanish automatically, simplifying the algebra substantially.

## 7.8 Rotational Kinetic Energy

**The intuition and derivation.** A spinning object clearly has kinetic energy — but expressing it in terms of angular quantities (rather than each individual bit of the object's own linear speed) makes it far more usable. Starting from ordinary kinetic energy for a single point mass at radius $r$, and substituting $v=r\omega$ (Unit 5.4):
$$K = \tfrac12mv^2 = \tfrac12m(r\omega)^2 = \tfrac12(mr^2)\omega^2 = \tfrac12I\omega^2$$
$$\boxed{K_\text{rotational} = \tfrac12I\omega^2}$$
— a genuinely direct parallel to $K=\frac12mv^2$, with $I$ standing in for $m$ and $\omega$ standing in for $v$, exactly the pattern that's been running through this whole unit.

**Combining translation and rotation — the general case.** An object like a rolling ball is simultaneously moving through space (its center of mass translating) *and* spinning about its own center — so it carries **both** kinds of kinetic energy at once:
$$K_\text{total} = \underbrace{\tfrac12mv_\text{cm}^2}_{\text{translational}} + \underbrace{\tfrac12I_\text{cm}\omega^2}_{\text{rotational}}$$
This is the *general* formula for the total kinetic energy of any rolling (or otherwise simultaneously translating-and-rotating) object.

---

## 7.9 Rolling Without Slipping

**The intuition.** "Rolling without slipping" means the object's rotation and its translation are locked together by a specific constraint: $v_\text{cm}=r\omega$ (and correspondingly $a_\text{cm}=r\alpha$) — the same relationship as Unit 5.4's $v=r\omega$, but now describing the object's *own* center-of-mass motion relative to its *own* spin, rather than a point moving around some external circle.

**Why this requires (static, not kinetic) friction.** Without *any* friction at all, a round object sliding down a frictionless incline would simply slide, never spinning up at all (there'd be no force applying any torque about its center). **Static friction** is what supplies the necessary torque to spin the object up as it moves — and it's specifically *static*, not kinetic, because (for true rolling without slipping) the contact point is, at each instant, momentarily at rest relative to the surface — there's no relative sliding at the contact point itself, even though the object as a whole is very much moving.

**A crucial energy consequence: static friction does *no* work.** Since work requires actual relative displacement in the direction of the force (Unit 3.1), and the contact point isn't sliding at all, static friction contributes zero work to the system — meaning **energy conservation still applies cleanly** to rolling-without-slipping problems, even though friction is technically present and technically necessary for the rolling to happen at all.

**Worked example — the classic race: a sliding block versus a rolling ball, both released from the same height on the same incline.** Which one reaches the bottom faster?

- **The sliding block** (frictionless incline, so it just slides, no rotation involved): all its gravitational PE converts entirely into *translational* KE: $mgh=\tfrac12mv^2 \Rightarrow v=\sqrt{2gh}$ — exactly the free-fall-equivalent speed from Unit 3.10.
- **The rolling ball:** now gravitational PE has to split between translational *and* rotational KE: $mgh = \tfrac12mv^2+\tfrac12I\omega^2$. Using $I=\frac25mR^2$ (solid sphere, 7.5) and $\omega=v/R$ (rolling constraint): $mgh=\tfrac12mv^2+\tfrac12(\tfrac25mR^2)(v/R)^2 = \tfrac12mv^2+\tfrac15mv^2 = \tfrac{7}{10}mv^2$, giving $v=\sqrt{\tfrac{10}{7}gh}$ — **noticeably smaller** than the sliding block's $\sqrt{2gh}$.

**The physical reason, worth stating plainly:** the rolling ball has to "spend" some of its available gravitational PE spinning itself up, leaving *less* left over for straight-line translational speed — so, released from the same height, a frictionless sliding block always beats a rolling ball to the bottom, and (following the same logic with the moment-of-inertia table from 7.5) a solid sphere would beat a hollow spherical shell, which would beat a disk, which would beat a ring — the *more* spread out an object's mass is from its own center, the more rotational KE it needs to "pay into," and the slower it reaches the bottom.

**Common mistake:** using only $\tfrac12mv^2$ for a rolling object's kinetic energy in an energy-conservation setup, forgetting the $\tfrac12I\omega^2$ rotational term entirely — an extremely common error specifically in "released from rest, rolls down an incline" problems.

## 7.10 Angular Momentum

**The intuition and derivation.** Just as impulse and momentum (Unit 4.1) came from combining $F=ma$ with a kinematics equation, **angular momentum** comes from combining the rotational analog of that same idea. Starting from the Impulse-Momentum Theorem, $F_\text{net}t=\Delta p$, and multiplying both sides by $r$ (the same trick used to derive $\tau=I\alpha$ in 7.3):
$$F_\text{net}tr = \Delta p \cdot r$$
Defining **angular momentum** as $L=rmv=rp$ for a single point mass (or, more generally and more usefully, $L=I\omega$ — which turns out to be correct for *any* rigid body, not just a single point mass):
$$\tau_\text{net}\cdot t = \Delta L$$
— the direct rotational parallel to the Impulse-Momentum Theorem, with torque playing the role of force, and angular momentum playing the role of ordinary momentum.

**A note on the two forms of $L$:** $L=rmv$ (or $L=rp$) is only valid for a single point mass moving in a circle. The more general form, $L=I\omega$, works for any rigid body (a spread-out object with its own moment of inertia $I$, spinning at angular velocity $\omega$) — and reduces exactly to $L=rmv$ for the special case of a single point mass (since $I=mr^2$ and $\omega=v/r$ there, giving $L=mr^2\cdot(v/r)=mrv$, matching).

---

## 7.11 Conservation of Angular Momentum

**The intuition.** If the net *external* torque on a system is zero, its total angular momentum never changes — the direct rotational parallel to conservation of (linear) momentum (Unit 4.4), and for the exact same underlying reason: internal torques, like internal forces, always come in equal-and-opposite pairs that cancel when the whole system is considered together.

**Worked example 1 — a single object, radius changing (directly from the handout).** A small block on a frictionless horizontal surface is attached to a cord passing through a hole in the surface, revolving at radius $0.30\ \text{m}$ with angular speed $2.00\ \text{rad/s}$. The cord is pulled from below, shortening the radius to $0.15\ \text{m}$.

**(a) Is angular momentum conserved here?** Yes — the tension in the cord always points directly toward the hole (the center of the circular path), meaning it has **zero lever arm** about that center, and therefore exerts **zero torque** about it (recall from 7.2: a force directed exactly along the line to the pivot produces zero torque, regardless of its size). With zero net external torque, $L$ is conserved throughout, even as the radius changes.

**(b) The new angular speed:** $L_1=L_2 \Rightarrow I_1\omega_1 = I_2\omega_2$. For a point mass, $I=mr^2$, so $mr_1^2\omega_1 = mr_2^2\omega_2$ (mass cancels): $\omega_2 = \omega_1(r_1/r_2)^2 = 2.00(0.30/0.15)^2 = 8.00\ \text{rad/s}$ — the block spins up considerably as it's pulled inward, even though no torque is doing the "spinning up" directly; it's angular momentum conservation, not a torque, causing the speed change.

**Worked example 2 — a collision between two spinning cylinders (directly from the handout).** A cylinder with moment of inertia $I_1$ spins at $\omega_i$ about a frictionless vertical axle. A second, initially non-spinning cylinder with moment of inertia $I_2$ drops onto the first; friction between the surfaces eventually brings them to a common final angular speed $\omega_f$.

**(a) Finding $\omega_f$:** the axle is frictionless (no *external* torque about that axis), and the friction between the two cylinders is entirely *internal* to the combined system — so angular momentum is conserved: $I_1\omega_i = (I_1+I_2)\omega_f \Rightarrow \omega_f = \dfrac{I_1\omega_i}{I_1+I_2}$.

**(b) Kinetic energy decreases:** compute $K_i=\tfrac12I_1\omega_i^2$ and $K_f=\tfrac12(I_1+I_2)\omega_f^2$; substituting the result from (a) and simplifying shows $K_f < K_i$ — **this is directly analogous to a perfectly inelastic linear collision** (Unit 4.7): angular momentum is conserved, but kinetic energy is not, because the two surfaces slide against each other (kinetic friction) before locking together, dissipating energy as heat, exactly the way two colliding, sticking masses lose translational KE.

**Worked example 3 — the spinning skater (a qualitative, but very physical, example).** A figure skater spinning with arms extended pulls them inward. Pulling the arms in *decreases* $I$ (since mass moves closer to the rotation axis, and $I$ depends on $r^2$) — with no external torque acting (ice is essentially frictionless, and gravity/normal-force act along the rotation axis, contributing no torque about it), $L=I\omega$ must stay constant, so $\omega$ must *increase* to compensate. This is exactly why skaters visibly spin faster the moment they pull their arms in, and why they slow back down when they extend their arms back out. **The skater's own angular speed reaches its maximum at exactly the moment their mass is most tightly gathered near the rotation axis** (arms and legs pulled in as close as possible), and its minimum when limbs are extended as far out as possible — a direct, visible illustration of $L=I\omega$ staying fixed while $I$ and $\omega$ trade off against each other.

**Common mistake:** assuming $\omega$ stays the same when $I$ changes (or vice versa). It's the **product** $L=I\omega$ that's conserved, not either individual factor — a shrinking $I$ always corresponds to a growing $\omega$ (and vice versa), never to $\omega$ staying put.

## 7.12 Comparing Translational and Rotational Motion — the Full Picture

It's worth laying the entire parallel out side by side now that every piece is in place — this table is, in a real sense, the summary of Units 1 through 4 *and* this entire unit at once:

| Quantity / Law | Translational | Rotational |
|---|---|---|
| Position-like quantity | $x$ | $\theta$ |
| Velocity-like quantity | $v$ | $\omega$ |
| Acceleration-like quantity | $a$ | $\alpha$ |
| "Inertia" quantity | $m$ | $I$ |
| Cause of acceleration | $F$ | $\tau$ |
| Newton's 2nd Law | $F_\text{net}=ma$ | $\tau_\text{net}=I\alpha$ |
| Kinetic energy | $K=\tfrac12mv^2$ | $K=\tfrac12I\omega^2$ |
| Momentum-like quantity | $p=mv$ | $L=I\omega$ |
| Impulse-momentum / Angular impulse-angular momentum | $Ft=\Delta p$ | $\tau t=\Delta L$ |
| Conservation law | conservation of momentum (no ext. force) | conservation of angular momentum (no ext. torque) |

**The single unifying idea behind this entire table:** every rotational law here was derived, not independently discovered, by taking an already-known translational law and substituting the rotational analog for each linear quantity — a pattern worth trusting and leaning on as a genuine memory and derivation shortcut, rather than treating rotational dynamics as an entirely separate body of knowledge to memorize from scratch.

---

## Unit 7 Summary

- **Torque** is the rotational analog of force, and depends on both the size of the applied force and *where* (and at what angle) it's applied — captured by $\tau=Fd\sin\theta$.
- **Moment of inertia** is the rotational analog of mass, but crucially depends on how mass is *distributed* relative to the rotation axis — the same object can have very different $I$ values about different axes.
- **$\tau_\text{net}=I\alpha$** (Newton's Second Law for rotation) and **$L=I\omega$** (angular momentum) both fall directly out of their translational counterparts, via the same "multiply by $r$" trick used throughout this unit.
- **Rolling without slipping** requires static friction (which does zero work), splits kinetic energy between translational and rotational forms, and explains why differently-shaped objects race down an incline at different speeds.
- **Conservation of angular momentum** applies whenever net external torque is zero — explaining the spinning skater, radius-changing orbits (tying back to Unit 6.8's Kepler's Second Law), and rotational "collisions."

**The single biggest meta-mistake in this unit:** trying to treat rotational dynamics as a completely new topic to memorize from scratch, rather than recognizing (and using) the direct, derivation-level parallel to translational dynamics from Units 1–4.

# UNIT 8 — Simple Harmonic Motion
### *(Handout, Class 13)*

This unit returns to the spring force first glimpsed back in Unit 3.3 ($F=kx$, and its associated potential energy $U=\tfrac12kx^2$) and builds an entire, self-contained topic around it: motion that endlessly repeats, driven by a force that always pulls back toward equilibrium.

---

## 8.1 What Makes Motion "Simple Harmonic"

**The intuition.** Simple Harmonic Motion (SHM) is defined by a very specific property of the force involved: it must be a **restoring force, directly proportional to displacement from equilibrium**, always pointing back toward that equilibrium. The farther you pull the system from equilibrium, the harder it's pulled back — and *exactly* proportionally, not just "roughly" or "eventually."

**The equation (Hooke's Law).** $F=-kx$, where $x$ is displacement from equilibrium and $k$ is a constant (the "spring constant," describing how stiff the spring is) — the minus sign is doing real conceptual work here, encoding "always pulls back toward $x=0$," regardless of which side of equilibrium the object currently sits on.

**A second, equivalent way to describe the same motion — directly, as a function of time.** Rather than only describing SHM through its force, it can be described directly as: $x(t) = A\cos(\omega t)$, where $A$ is the amplitude (the maximum displacement reached) and $\omega$ here is an **angular frequency** describing how rapidly the oscillation repeats (worth connecting back to the subtlety flagged in Unit 5.3 — this "$\omega$" is not describing any actual physical rotation happening here; it's borrowed notation, because the mathematics of $\cos(\omega t)$ genuinely behaves the same way whether it's describing literal circular motion or this back-and-forth oscillation).

**Why these two descriptions (the force law, and the position-vs-time formula) are equivalent** is worth understanding conceptually, even without formal calculus: a restoring force proportional to displacement is *exactly* the condition that produces smooth, repeating, cosine-shaped motion — pull it far away, it gets pulled back hard and overshoots; the overshoot on the other side then gets pulled back hard too; and this back-and-forth naturally traces out the same repeating cosine shape every single cycle, forever (in the idealized, frictionless case).

**Worked example — recognizing SHM, versus not.** Stretch a spring twice as far, and (by $F=-kx$) it pulls back with exactly twice the force — a clean, direct proportionality, confirming SHM. **Contrast this with a ball bouncing on a hard floor:** it does oscillate up and down repeatedly, but the force involved (gravity while airborne, then a sudden, very different impact force at the bounce) is **not** proportional to displacement from some equilibrium point in the SHM sense at all — so a bouncing ball, despite superficially "oscillating," is not an example of Simple Harmonic Motion.

**Common mistake:** assuming any repetitive, back-and-forth motion automatically qualifies as SHM. The specific requirement — a restoring force *directly proportional* to displacement — is what actually defines the category, and plenty of everyday repetitive motions (a bouncing ball, a person walking) don't satisfy it.

## 8.2 Potential Energy in SHM

**The intuition and derivation, mirroring Unit 3.6's introduction of potential energy exactly.** Since $F=-kx$ is a conservative force (its associated work only depends on start and end position, not on path — the same property gravity has), it has an associated potential energy, derivable the same way as before: $W=-\Delta U_s$, and since the work done by a variable force is the area under the F-x graph (Unit 3.3) — here a straight line through the origin, so the "area" is a triangle:
$$U_s = \tfrac12kx^2$$
— exactly the formula previewed back in Unit 3.3, now given its proper home as the defining potential energy of this entire unit.

---

## 8.3 The Motion Over Time — Position, Velocity, and Acceleration Graphs

**The intuition.** Given $x(t)=A\cos(\omega t)$ (8.1), the velocity and acceleration follow by the exact same "rate of change" logic used throughout this course (Units 1, 5, 7) — even without formally invoking calculus, it's worth simply stating the results and reading them graphically:

$$x(t) = A\cos(\omega t) \qquad v(t) = -\omega A\sin(\omega t) \qquad a(t) = -\omega^2A\cos(\omega t) = -\omega^2x(t)$$

**Reading these three graphs together, as a set (all sharing the same period $T$):**
- **The $x$-$t$ graph** starts at its maximum ($A$, since $\cos(0)=1$), decreases smoothly through zero, reaches $-A$, and returns — a smooth, symmetric wave.
- **The $v$-$t$ graph** starts at **zero** (since $\sin(0)=0$) — the object is momentarily at rest right when it's at maximum displacement — then reaches its own maximum magnitude a quarter-period later, exactly when $x$ crosses through zero.
- **The $a$-$t$ graph** mirrors the $x$-$t$ graph exactly, just flipped upside down and scaled by $\omega^2$ — directly visible in the relation $a=-\omega^2x$: **acceleration is always exactly proportional to (and opposite in sign from) displacement**, the same restoring-force idea from 8.1, now expressed directly as a statement about acceleration rather than force.

**A genuinely useful pattern-recognition fact:** whenever you see $a=-(\text{positive constant})\times x$ show up in *any* new context (not just springs), you can immediately recognize SHM is present, and immediately identify that constant as $\omega^2$ — a shortcut worth having ready, since it lets you spot SHM even in unfamiliar-looking setups.

---

## 8.4 Why the Ordinary Kinematics Formulas No Longer Apply

**The intuition, and an important warning worth flagging explicitly.** The four kinematics formulas from Unit 1.6 were derived specifically for **constant acceleration**. In SHM, acceleration is emphatically *not* constant — it constantly changes as $x$ itself changes (following $a=-\omega^2x$ from 8.3) — so **none of the Unit 1 kinematics formulas can be used here.** This is a genuinely different mathematical regime, requiring the specific relationships in 8.3 and 8.5 instead.

**What connects the different SHM quantities, instead of the old kinematics formulas:**
- Acceleration is related to force (via $F=ma=-kx$, and $a=-\omega^2x$).
- Velocity is related to kinetic energy (via $K=\tfrac12mv^2$, connecting to 8.5 next).
- Potential energy is related to position (via $U=\tfrac12kx^2$, from 8.2).

---

## 8.5 Energy in SHM — the Constant Trade-Off Between Kinetic and Potential

**The intuition.** Exactly as in Unit 3.10's conservation of mechanical energy, an oscillator with no friction conserves its *total* mechanical energy throughout the motion — but here, that energy is continuously and smoothly trading back and forth between kinetic and potential forms as the object moves.

**The equation.** $E = K+U = \tfrac12mv^2+\tfrac12kx^2 = \text{constant}$, and this constant total can be expressed cleanly using the amplitude: $E=\tfrac12kA^2$ (evaluating at the extreme point $x=A$, where $v=0$ momentarily, so *all* the energy is potential at that instant).

**Reading the graphs (directly connecting to 8.3's velocity/position graphs):**
- **At $x=0$ (equilibrium):** $U=0$ entirely, so *all* the energy is kinetic — this is exactly where speed is at its **maximum**, matching the $v$-$t$ graph from 8.3 peaking right when $x$ crosses zero.
- **At $x=\pm A$ (the extremes):** $v=0$ entirely, so *all* the energy is potential — the object is momentarily motionless at the very edges of its swing.
- **In between:** energy continuously splits between the two forms, always summing to the same constant $E$.

**A genuinely counterintuitive fact worth internalizing carefully — resolving the Concept Check from 8.1:** speed is **maximum** exactly where the restoring force (and thus acceleration) is **zero** (at equilibrium), and speed is **zero** exactly where the restoring force is at its **maximum** (at the extremes). These two quantities are, in a real sense, precisely out of step with each other throughout the motion — many students' first instinct is to expect speed to be largest where the "push" is strongest, but SHM's specific mechanics works the opposite way: the strong restoring force at the extremes is exactly what has spent the whole *previous* quarter-cycle decelerating the object down to zero speed there, while the object cruises through equilibrium at top speed, precisely because there's momentarily no restoring force acting on it there at all.

## 8.6 The Period of a Spring-Mass System

**The intuition.** How long one full oscillation cycle takes turns out to depend on exactly two things: how stiff the spring is, and how much mass is being pushed around by it — in a specific, worth-remembering combination.

**The equation.** $f=1/T$, $\omega=2\pi f$ (the same period/frequency/angular-frequency relationships from Unit 5.3, now applied in this new context — again, a notational borrowing, not literal rotation), and specifically for a spring-mass system:
$$T = 2\pi\sqrt{\frac{m}{k}}$$

**Building intuition for *why* it looks this way, without a full derivation:** a **stiffer** spring (larger $k$) pulls back harder for the same displacement, completing each cycle **faster** — hence $k$ sits in the denominator (larger $k$, smaller $T$). A **heavier** mass (larger $m$) has more inertia to overcome each cycle, taking **longer** — hence $m$ sits in the numerator (larger $m$, larger $T$).

**A genuinely surprising and important fact, worth sitting with:** the period does **not** depend on amplitude $A$ at all. A gentle, small-amplitude oscillation and a large, vigorous one (same spring, same mass) take exactly the same time to complete one cycle — amplitude affects how *far* and how *fast* the mass swings at each point, but never *how long* a cycle takes.

**Common mistake:** assuming a larger-amplitude oscillation (stretched further before release) must complete its cycle more slowly (more distance to cover) or more quickly (more restoring force available). Neither is true — amplitude simply doesn't appear in the period formula at all.

---

## 8.7 The Pendulum — a Second Example of SHM

**The intuition.** A pendulum swinging through a **small** angle also behaves as SHM, with gravity supplying the restoring force (rather than a spring's elastic force) — the tangential component of gravity along the pendulum's arc grows roughly proportionally to displacement, for small swings specifically, which is exactly the condition needed for SHM (8.1).

**The equation.**
$$T = 2\pi\sqrt{\frac{L}{g}}$$
valid specifically for **small angles** (roughly under $15°$) — for larger swings, the restoring force is no longer cleanly proportional to displacement, and the motion, while still periodic, is no longer *simple* harmonic motion in the strict sense, and this formula stops being accurate.

**A remarkable and genuinely useful fact: mass doesn't appear anywhere in this formula.** A heavy bob and a light bob, on the same length string, swing with *exactly* the same period — for the same underlying reason a heavier object doesn't fall faster in free fall (Unit 1.8): gravity's pull scales with mass exactly the same way inertia (resistance to acceleration) does, so mass cancels out of the equation of motion entirely.

**Some genuinely practical applications, worth thinking through:**
- **Clocks.** A pendulum clock's timing is set entirely by the pendulum's *length* — a clock running slow is fixed by adjusting length, never by changing the (irrelevant) mass of the bob.
- **Effect of moving to the Moon.** Since $T\propto1/\sqrt{g}$, and the Moon's gravity is weaker than Earth's, a pendulum clock brought to the Moon would run **slower** (longer period for the same length) — ticking less often per actual elapsed time, and so under-counting time if it's still being read as though it were on Earth.
- **Measuring the height of a tall structure.** A pendulum's period depends only on its length and $g$ — so, conversely, *measuring* a pendulum's period lets you solve for its length precisely, a genuinely used historical technique (hang a pendulum from the top of a tower, time its swing, solve $T=2\pi\sqrt{L/g}$ for $L$, and you've measured the tower's height without ever climbing up with a tape measure).

**Common mistake:** assuming a heavier pendulum bob swings with a different (often assumed "slower," from an intuitive sense of heavier things being sluggish) period than a lighter one. Mass simply doesn't appear in the formula — only length and $g$ matter, for small-angle swings.

## 8.8 The Vertical Spring-Mass System

**The intuition.** Everything above was implicitly framed for a horizontal spring, where gravity plays no role in the oscillation at all. Hang the same spring **vertically** with a mass attached, and gravity now constantly pulls the mass downward, in addition to the spring's own restoring force — but the resulting motion is, remarkably, *still* simple harmonic, just oscillating about a **new** equilibrium position, shifted downward from the spring's natural (unstretched) length.

**Finding the new equilibrium.** At the new balance point, the spring's own restoring force upward exactly cancels gravity: $kd=mg\Rightarrow d=mg/k$ (where $d$ is how much the spring has stretched from its natural length, just to reach the new resting position, before any oscillation even begins).

**Why the motion around this new equilibrium is still ordinary SHM, with the same period formula.** If you measure displacement $y$ *relative to this new equilibrium point* (rather than relative to the spring's unstretched natural length), the net force turns out to still be exactly $-ky$ (the constant gravity term and the constant equilibrium-stretch term cancel each other out identically, leaving only a term proportional to displacement from the *new* equilibrium) — meaning the vertical spring-mass system oscillates with **exactly the same period formula**, $T=2\pi\sqrt{m/k}$, as the horizontal case; gravity only shifts *where* the oscillation is centered, never *how fast* it oscillates.

**A practical application: the simple bathroom-style scale.** A scale that measures weight by how far a spring compresses under a load is a direct, everyday application of $kd=mg$ from above — measure the compression $d$, and (knowing the scale's spring constant $k$) directly compute the weight $mg$ pressing down on it.

## Unit 8 Summary

- **SHM is defined by a restoring force directly proportional to displacement** ($F=-kx$), which produces the characteristic $x(t)=A\cos(\omega t)$ motion.
- **The ordinary kinematics formulas (Unit 1) do not apply** here, since acceleration is not constant — instead, use the specific SHM relationships between force/acceleration/position, velocity/kinetic energy, and potential energy/position.
- **Speed is maximum at equilibrium and zero at the extremes** — the reverse of where the restoring force and acceleration are strongest, a genuinely counterintuitive but central fact of SHM.
- **Period depends on $m$ and $k$ (springs) or $L$ and $g$ (pendulums) — never on amplitude, and never on mass for a pendulum.**
- **A vertical spring-mass system** oscillates about a new, gravity-shifted equilibrium point, but with the *same* period as the horizontal case.

**The single biggest meta-mistake in this unit:** reaching for the Unit 1 kinematics formulas out of habit, when the non-constant acceleration in SHM makes them invalid — use the force/energy relationships specific to this unit instead.

# UNIT 9 — Fluid Mechanics
### *(Handout, Class 14 — the final unit of the course)*

Every previous unit dealt with discrete, well-defined objects — blocks, balls, satellites. Fluids (liquids and gases) don't hold a fixed shape at all, which means force and mass, the tools we've used all course, need new forms — pressure instead of force, density instead of mass — before this unit's tools (buoyancy, flow, Bernoulli's equation) become usable.

---

## 9.1 Solids vs. Fluids, and How to Describe Them

**The intuition.** A solid holds both its own shape and its own volume. A fluid (this category includes *both* liquids and gases) holds its volume (for a liquid) or not even that (for a gas), but never holds a fixed shape — it simply takes the shape of whatever container holds it. This single difference is exactly why fluids need a different descriptive toolkit than the rigid blocks and point masses of earlier units.

**Density, replacing mass.** Since a fluid doesn't have a fixed shape or a naturally defined "one object" to weigh, it's far more useful to describe *how much mass is packed into a given volume*:
$$\rho = \frac{M}{V}$$
**Specific gravity**, a convenient dimensionless comparison: $\text{sp. gr.} = \rho_\text{material}/\rho_\text{water}$, using water's density ($1000\ \text{kg/m}^3$) as a universal reference point — a material with specific gravity $2.0$ is simply "twice as dense as water," without needing to remember or restate the actual density number each time.

**Pressure, replacing force.** Since a fluid pushes on *every* bit of surface it touches, not just one localized contact point, it's more useful to describe *how much force is spread across a given area*:
$$P = \frac{F_\perp}{A}$$
Unit: the **Pascal** (Pa), $\text{N/m}^2$ — worth noting immediately that a Pascal is a genuinely *small* unit in everyday terms: atmospheric pressure itself is about $P_\text{atm}=1.013\times10^5\ \text{Pa}$, and (as a nice concrete sense of scale) even the weight of a handful of flour ($0.1\ \text{kg}$, about $1\ \text{N}$ of weight) spread over a full $1\ \text{m}\times1\ \text{m}$ table only amounts to $1\ \text{Pa}$ of pressure — a genuinely tiny amount, dwarfed by ordinary atmospheric pressure pressing on everything around us all the time.

**Gauge pressure.** $P_\text{gauge} = P_\text{total} - P_\text{atm}$ — many everyday pressure measurements (tire pressure, blood pressure) are actually reported as *gauge* pressure, i.e., how much *above* atmospheric the true pressure is, rather than the full absolute value — worth keeping in mind when interpreting a stated pressure reading.

---

## 9.2 The Pressure of a Liquid — Why It Increases With Depth

**The intuition, and a simple derivation.** Imagine a thin, cylindrical column of liquid, extending from the surface down to some depth $h$. The liquid at the very bottom of this column has to support the weight of *all* the liquid sitting on top of it, in addition to whatever pressure is being applied at the surface (atmospheric pressure, typically). Balancing the forces on this imaginary column (top pressure pushes down, bottom pressure pushes up, and the column's own weight pulls down):
$$P_\text{bottom}\cdot A = P_\text{top}\cdot A + (\text{weight of the column})= P_0 A + \rho(Ah)g$$
Dividing through by $A$:
$$P_\text{bottom} = P_0 + \rho gh \qquad\Rightarrow\qquad P_\text{fluid} = P_\text{bottom}-P_0 = \rho gh$$
**The key variable here is depth, $h$** — pressure depends only on how far below the surface you are, not on the shape of the container, or how much total liquid surrounds you at that depth.

**Pascal's Principle.** If an external pressure is applied *anywhere* to an enclosed fluid, that added pressure is transmitted **uniformly** to every part of the fluid — this is the working principle behind hydraulic systems (car lifts, hydraulic brakes): a small applied force over a small area creates a pressure that's transmitted, unchanged, to a much larger area elsewhere in the connected fluid, producing a correspondingly much larger force there.

## 9.3 Measuring Blood Pressure — a Real Application

**The intuition.** Blood pressure is traditionally reported in **mm Hg** (millimeters of mercury) — literally, the height of a mercury column that the actual blood pressure would be able to support, via exactly the $P=\rho gh$ relationship from 9.2. Mercury (specific gravity $13.6$ — over 13 times denser than water) is used specifically *because* it's so dense: it lets a pressure be represented by a conveniently small, easily-readable column height, rather than requiring an unwieldy multi-meter-tall column of water to represent the same pressure.

**Reference values worth knowing:** normal systolic pressure (the peak, as the heart contracts) is under about 120 mm Hg; normal diastolic pressure (the resting value, between heartbeats) is under about 80 mm Hg — hence the familiar "120/80" blood pressure reading.

**A related unit, for contrast:** tire pressure is often reported in **psi** (pounds per square inch) — a completely different, non-SI unit system, worth recognizing on sight so as not to confuse it with mm Hg or Pascals when reading a real-world specification.

---

## 9.4 The Buoyant Force — Archimedes' Principle

**The intuition and derivation, connecting directly back to 9.2.** An object submerged in a fluid has fluid pressing on it from every direction — but since pressure *increases* with depth (9.2), the pressure pushing **up** on the object's bottom surface is always *greater* than the pressure pushing **down** on its top surface (the bottom is deeper, and therefore at higher pressure, than the top). This pressure *difference*, multiplied by the object's cross-sectional area, produces a net **upward** force — the buoyant force.

**The equation (Archimedes' Principle).**
$$F_\text{buoy} = \rho_\text{fluid}V_\text{sub}g$$
where $V_\text{sub}$ is specifically the **submerged** volume (the volume of fluid actually displaced) — worth noting this holds true **regardless of whether the object is fully or only partially submerged**; the formula only ever cares about the volume of fluid genuinely displaced, not the object's total volume.

**A useful way to remember and interpret this formula:** the buoyant force is always exactly equal to **the weight of the fluid that the object has displaced** — a submerged rock pushes aside (displaces) a certain volume of water, and the buoyant force pushing that rock upward is exactly the weight that same volume of water would have had, on its own.

---

## 9.5 Floating, Sinking, and Neutral Buoyancy

**The intuition.** If an object floats, it's in equilibrium (Unit 2.18's logic, applied vertically): the buoyant force exactly balances its weight.

**Deriving the floating condition.** $mg = \rho_\text{fluid}V_\text{sub}g \Rightarrow \rho_\text{object}Vg = \rho_\text{fluid}V_\text{sub}g$ (using $m=\rho_\text{object}V$, the object's *total* volume $V$, not just the submerged part) $\Rightarrow$
$$\frac{V_\text{sub}}{V} = \frac{\rho_\text{object}}{\rho_\text{fluid}} = \text{sp. gr.}$$
— a genuinely elegant result: **the fraction of an object's volume that ends up submerged is exactly equal to its specific gravity** (relative to that fluid).

- **If sp. gr. $<1$:** $V_\text{sub}/V<1$ means only *part* of the object needs to be submerged to displace enough fluid to support its weight — the object **floats**, with the rest sticking out above the surface (ice, at roughly 0.92 specific gravity relative to water, floats with about 92% of its volume submerged and 8% visible above the surface).
- **If sp. gr. $=1$:** $V_\text{sub}/V=1$ — the *entire* object must be submerged (but no more is needed, and no more is possible than 100%) to balance — the object is **neutrally buoyant**, staying wherever placed within the fluid, neither rising nor sinking.
- **If sp. gr. $>1$:** the equation would require $V_\text{sub}/V>1$ — displacing *more* volume than the object even has — which is physically impossible (an object can't submerge more than its own full volume). **This is exactly why such an object sinks**: it's fundamentally unable to displace enough fluid, at any depth, to generate a buoyant force large enough to match its own weight, so gravity wins and it sinks all the way down.

**Common mistake:** trying to apply the equal-volumes floating equation ($mg=\rho_\text{fluid}Vg$, using the object's *full* volume) to a sinking object. That equation is only valid for the *floating* case (where $V_\text{sub}$ is genuinely less than or equal to $V$) — a sinking object never reaches force balance at all; it simply accelerates downward until it hits the bottom of the container.

## 9.6 The Equation of Continuity

**The intuition.** For an incompressible fluid (essentially all liquids, to a very good approximation) flowing through a pipe, the *same* amount of volume has to pass every cross-section of the pipe per unit time — fluid can't pile up or vanish partway through. If the pipe narrows, the fluid has no choice but to **speed up** to keep the same volume moving through per second; if it widens, the fluid **slows down**.

**The equation.** $A_1v_1 = A_2v_2$, i.e., $Av=\text{constant}$ throughout the pipe (the "volume flow rate," literally cross-sectional area times speed, stays fixed).

**Everyday intuition, worth connecting explicitly:** this is exactly why putting your thumb partially over a garden hose's opening (reducing the effective area $A$) makes the water shoot out **faster** — the same total volume of water has to pass through a smaller opening in the same time, so its speed must increase to compensate.

---

## 9.7 Bernoulli's Equation

**The intuition — energy conservation, translated into fluid language.** Bernoulli's Equation is, at its core, the Unit 3 conservation-of-energy idea ($E=K+U$, staying constant when no non-conservative force does work), rewritten in terms of pressure, height, and flow speed instead of an individual object's kinetic and potential energy.

**Viscosity — the fluid version of friction.** Viscosity is a fluid's internal "friction" — resistance to flow, caused by the fluid's own molecules rubbing against each other as they move. Thick, syrupy fluids like ketchup have high viscosity; thin fluids like water have comparatively low viscosity. **Bernoulli's Equation specifically assumes viscosity is negligible** — meaning no mechanical energy is lost to internal friction as the fluid flows, mirroring the frictionless assumption behind Unit 3.10's simple energy conservation.

**The equation.**
$$P_1+\rho gh_1+\tfrac12\rho v_1^2 = P_2+\rho gh_2+\tfrac12\rho v_2^2 \qquad\text{i.e.}\qquad P+\rho gh+\tfrac12\rho v^2 = \text{constant}$$
**Reading each term against its Unit 3 counterpart:** $\rho gh$ plays exactly the role of gravitational potential energy $mgh$ (but per unit *volume* rather than per object — using $\rho$, density, in place of $m$, mass); $\tfrac12\rho v^2$ plays exactly the role of kinetic energy $\tfrac12mv^2$, again per unit volume. The pressure term $P$ is the genuinely new piece, representing energy associated with the fluid's own internal pressure, not present in the ordinary mechanics energy equation at all.

---

## 9.8 Torricelli's Theorem — an Application of Bernoulli's Equation

**The intuition.** Consider a large, open container with a small hole near the bottom, some depth $h$ below the surface. Applying Bernoulli's Equation between the (essentially stationary, since the container is large) surface and the small hole (both open to atmospheric pressure, so the $P$ terms cancel from both sides):
$$v = \sqrt{2gh}$$
**This is exactly the same formula as an object in free fall from height $h$** (Unit 1.8, or Unit 3.10's energy-conservation shortcut) — the fluid exiting the hole moves at precisely the speed it would have reached by simply falling that same height $h$, a genuinely satisfying and memorable connection back to the very first unit of the course.

## 9.9 The Bernoulli Effect and the Venturi Meter

**The intuition — the genuinely counterintuitive headline result.** Consider fluid flowing horizontally through a pipe that narrows from a wide section (radius $R_1$) to a narrow section (radius $R_2 < R_1$), with $h_1=h_2$ (purely horizontal, so the height terms in Bernoulli's equation cancel and can be ignored entirely).

**Combining the two tools from this unit.** From the Equation of Continuity (9.6), since $A_1>A_2$ (wide to narrow): $v_1<v_2$ — the fluid speeds up as it enters the narrow section. Now applying Bernoulli's Equation (9.7), with the height terms dropped:
$$P_1+\tfrac12\rho v_1^2 = P_2+\tfrac12\rho v_2^2$$
Since $v_1<v_2$, the kinetic-energy-like term is larger on the right side — which, for the equation to balance, forces $P_1>P_2$.

**The headline result, worth stating on its own: "the pressure is lower where the flow speed is greater."** This is, at first glance, quite counterintuitive — you might expect *faster*-moving fluid to "push harder," implying higher pressure — but Bernoulli's Equation shows the opposite: speeding up requires "spending" some of the fluid's pressure-energy to do so, converting it into the kinetic-energy-like term instead.

**The Venturi meter — a practical device built directly on this effect.** Two vertical open tubes, inserted into a pipe at a wide section and a narrow section respectively, show *different* liquid heights — the height difference directly reveals the pressure difference between the two points (via the ordinary $P=\rho gh$ relationship from 9.2, now applied to these small vertical measuring tubes, which are governed by ordinary fluid *statics*, even while the *horizontal* flow through the main pipe is governed by fluid *dynamics*). Measuring that height difference lets you back out the fluid's actual flow speed — a genuinely practical device for measuring flow rate without needing to interrupt or directly probe the flow itself.

**A real-world application worth knowing: how airplane wings generate lift.** A wing's shape is designed so air flows faster over the curved top surface than along the flatter bottom surface. By the exact same Bernoulli-Effect logic just derived, faster flow on top means **lower** pressure on top — and with higher pressure underneath than above, the wing experiences a net **upward** force: lift. (In reality, a fully accurate account of lift is more subtle and involves additional aerodynamic effects beyond this simplified picture — but the Bernoulli-based explanation captures a genuinely real part of the story, and is the standard first-pass explanation given at this course's level.)

## 9.10 Summary — the Four Master Formulas of Fluid Mechanics

The entire unit compresses down to four core relationships, each one worth being able to state and use on sight:

$$P_\text{fluid} = \rho gh \qquad F_\text{buoy} = \rho_\text{fluid}V_\text{sub}g \qquad Av = \text{constant} \qquad P+\rho gh+\tfrac12\rho v^2 = \text{constant}$$

**How they relate to each other, as a final unifying thought:** the first is fluid *statics* (nothing moving) — pressure building with depth. The second is a direct consequence of the first, applied to an object sitting in that pressure gradient. The third and fourth are fluid *dynamics* (fluid actually flowing), with the fourth (Bernoulli's Equation) being, in essence, this entire unit's version of energy conservation — and, not coincidentally, reducing to the *first* formula exactly whenever the flow speed is zero everywhere ($v_1=v_2=0$), tying the whole unit back together as one consistent framework rather than four disconnected facts.

## Unit 9 Summary

- **Fluids require pressure (not force) and density (not mass)** as their basic descriptive quantities, since a fluid has no single fixed shape or well-defined "one object."
- **Pressure increases linearly with depth** ($P=\rho gh$), which directly explains buoyancy: greater pressure pushing up from below than pushing down from above creates a net upward force.
- **Archimedes' Principle** ($F_\text{buoy}=\rho_\text{fluid}V_\text{sub}g$) determines floating vs. sinking through a genuinely elegant result: the submerged fraction of a floating object exactly equals its specific gravity.
- **The Equation of Continuity** ($Av=\text{const}$) and **Bernoulli's Equation** together describe flowing fluids, with Bernoulli's Equation serving as this unit's version of energy conservation.
- **The Bernoulli Effect** ("pressure is lower where flow speed is greater") explains the Venturi meter and (approximately) how airplane wings generate lift.

**The single biggest meta-mistake in this unit:** trying to apply the simple floating-equilibrium equation to an object that's actually sinking (where no such equilibrium exists), or forgetting that Bernoulli's Equation, like Unit 3's energy conservation, only applies cleanly when viscosity (the fluid equivalent of friction) can be neglected.
