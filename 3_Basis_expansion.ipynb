```python
import numpy as np
X = np.sort(10 * np.random.rand(200, 1), axis=0)
y = np.sin(X).ravel() + np.sin(2 * X).ravel() + np.random.normal(0, 0.3, X.shape[0])
```

### 1a. Find the optimal linear fit for this dataset
Plot both the data and the linear fit

```python
import matplotlib.pyplot as plt

col1 = np.ones((200,1))
col2 = X
col3 = np.sin(X)
col4 = np.sin(2*X)

phi = np.concatenate([col1,col2,col3,col4], axis = 1)

w = np.linalg.inv(phi.T @ phi) @ phi.T @ y

yp = phi @ w



plt.scatter(X,y)
plt.plot(X, yp)
```
