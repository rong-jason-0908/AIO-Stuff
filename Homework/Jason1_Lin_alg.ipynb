## Question 1:
Prove that for any two eigen vectors with different eigenvalues ($\lambda_1 \not = \lambda_2$), they are orthogonal (i.e. $v_1 ^T v_2 = 0$).    A is symmetric

We want to get $\lambda$ into an equation with $v_1$ and $v_2$ somehow, so we first take this:
$$
v_1^T A v_2
$$
and we rewrite in two different ways:
\
note that:
$$
A = A^T
$$
\
1st:
Since $v_1$ and $v_2$ are both eigen vectors, applying matrix $A$ is the same as using the eigen value to scale, which allows us to switch in $\lambda$
$$
v_1^TA^Tv_2 = (v_1^T A) v_2= (v_1^T A^T)v_2 = (A v_1)^Tv_2 = (\lambda_1 v_1)^T v_2=\lambda_1 v_1^Tv_2
$$

from the same equation, we can rewrite into the 2nd way:
$$
v_1^T A v_2 = v_1^T \lambda_2 v_2 = (\lambda_2)(v_1^T v_2)^T
$$
(we can transpose since the eigen vectors are of dimension 1xN; this means multiplying one by the transpose of the other creates a single 1x1, so our outer transpose does nothing)
so our equation turns into:
$$
(\lambda_2)(v_1^T v_2)^T = \lambda_2v_2^T v_1
$$
since both of these are derived from the same thing, we can state:

$$
\lambda_1 v_1^Tv_2 = \lambda_2v_2^T v_1
$$

We know that $\lambda_1 \ne \lambda_2$; but we also know that $v_1^T v_2 = v_2^T v_1$. (same values are multiplied)

From these two statements, we can state that the only possible value of $v_1^T v_2 = v_2^T v_1$ is $0$

Written out, we can do:

$$
(\lambda_1 - \lambda_2)(v_1^T v_2) = 0
$$
and since we know $\lambda$ s arent equal, $v_1^Tv_2$ must equal $0$

## Question 2: 
Given some vector $\phi(x) \in R^{1 \times N}$, which can be rewritten as $\phi(x) = U\Sigma V^T$, and matrix $A = \phi(x)^T \phi(x)$, prove the following:

1. The determinant of A is equal to the product of the eigenvalues of A.

The way A is built, we can assume it's symmetric (and therefore diagonalizable)

Continuing, we can split A into its spectral decomposition:

$$
A = Q \Lambda Q^T
$$

a property of the determinant is that for some matrix $A$ , we can set it equal to the product of the determinants of its decomposed parts (whether that be SVD or eigen decomp)

so:

$$
\det(A) = \det(Q) \det(\Lambda) \det(Q^T)
$$
since these are now constants, we can shuffle. In addition, we know that $\det(A) \times \det(B) = \det(A \times B)$
so our equation can be reformed into:

$$
\det(A)= \det(Q) \det(Q^T) \det(\Lambda) = \det(Q^T Q)\det(\Lambda)
$$
So now, we deal with $Q^T Q$ .

$Q$ is comprised of the orthonormal eigen vectors (each column as a vector); we know that multiplying some vector $v$ with its own transpose will give the same vector with magnitude squared.

for example:
$$
A = \begin{bmatrix} a & b \end{bmatrix}
$$

so 
$$
A \cdot A^T = \begin{bmatrix} a & b \end{bmatrix} \cdot \begin{bmatrix} a \\ b \end{bmatrix} = a^2 + b^2
$$
this operation is essentially the dot product.
Returning to our $QQ^T$:


$Q$ is comprised of orthonormal eigen vectors along its columns. so:

 $$Q^TQ = \begin{bmatrix} v_1^Tv_1 & v_1^Tv_2 & \dots \\ v_2^Tv1 & v_2^T v_2 & \dots \\ \vdots & \vdots & \ddots \end{bmatrix} $$

and now we can see what happens in this product:

since all $v_i$ are orthonormal, 2 things happen: off diagonal products become zero (dot product of perpendicular vectors is always 0) and diagonal products become 1 (since its magnitude squared, where magnitude = 1)

This goes by another name: the identity. 

so:

$$
Q^TQ = I
$$

the determinant of the identity is just 1. so our ending equation becomes:

$$
\det(A)= \det(Q) \det(Q^T) \det(\Lambda) = \det(Q^T Q)\det(\Lambda) = \det(\Lambda)
$$
We know that $\Lambda$ just has all the eigen values on its diagonal. We can do the last part of this proof geometrically. For some square matrix where there are values only along the diagonal, we can see that it purely scales for each dimension. We also know that the determinant is just the scaling factor. It follows logically that the determinant is the product of these scaling values, eg. the eigen values.

the first thing to show is that trace inputs can be cycled:

We know that $\operatorname{tr}(AB) = \operatorname{tr}(BA)$

\operatorname{tr}(XYZ) = \operatorname{tr}((XY)Z) = \operatorname{tr}(Z(XY))

so now we plug in A:
$$
\operatorname{tr}(A) = \operatorname{tr}(Q \Lambda Q^T) = \operatorname{tr}(Q^T Q \Lambda) 
$$
from the past proof, we know $Q^T Q = I$
so our equation becomes:
$$
\operatorname{tr}(\Lambda)
$$
trace is the sum of all values on the diagonal, which happen to be the eigen values of A.
